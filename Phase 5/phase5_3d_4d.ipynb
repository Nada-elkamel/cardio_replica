```python
# Generated from: el-final-mtaa-p5 (3).ipynb
# Converted at: 2026-08-06T21:19:08.350Z
# Next step (optional): refactor into modules & generate tests with RunCell
# Quick start: pip install runcell

# Quick start: pip install runcell

!pip install ukb-atlas

# ==============================================================================
# Phase 5 — Jumeau numérique cardiaque 3D/4D (VG/VD) avec référence saine
# ------------------------------------------------------------------------------
# Version nettoyée et dédupliquée du notebook original, avec RÉINTÉGRATION de
# l'étude d'ablation (AUC/Calibration ECG-only vs ECG+Tabulaire vs
# ECG+Graph+Tabulaire), qui avait été retirée par erreur.
#
# Ce que ce script fait, dans l'ordre :
#   1. Imports & configuration
#   2. Chargement des données test + prédiction de probabilité LVEF≤45%
#   3. Reconstruction du maillage bi-ventriculaire (atlas UKB) pour un patient
#   4. Rendu 3D/4D UNIFIÉ : surface patient colorée + contour blanc de la
#      géométrie saine de référence, en un seul appel (render_biv_mesh)
#   5. Jumeau numérique 4D animé (cycle cardiaque ED -> ES -> ED)
#   6. Figure de comparaison de 3 patients (sain / IC modérée / IC élevée)
#   7. Étude d'ablation (apport du Graph Enrichment) — RÉINTÉGRÉE
#
# NB : le dashboard Streamlit et le pipeline ECG->graph embeddings du notebook
# d'origine restent hors du périmètre de ce script et ne sont pas réintégrés
# ici — seule l'étude d'ablation, explicitement demandée, a été remise.
# ==============================================================================

# ============================== 1. IMPORTS ===================================
import os
import random
import warnings
from functools import lru_cache
from typing import Dict, Iterable, Optional

import numpy as np
import pandas as pd
import h5py
import joblib
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers

from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, brier_score_loss, roc_curve
from sklearn.calibration import calibration_curve

import matplotlib
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import matplotlib.cm as cm
from matplotlib.colors import Normalize, to_rgb
from matplotlib.gridspec import GridSpec
from matplotlib.patches import Patch
from mpl_toolkits.mplot3d.art3d import Poly3DCollection, Line3DCollection

from scipy import sparse
from scipy.signal import find_peaks

try:
    from ukb import atlas as ukb_atlas
    from ukb import surface as ukb_surface
except ImportError as e:
    raise ImportError(
        "Le package 'ukb-atlas' est requis.\n"
        "Dans Kaggle : Settings > Internet > On, puis : !pip install ukb-atlas"
    ) from e


# ============================== 2. CONFIGURATION ==============================
KAGGLE_INPUT = "/kaggle/input/datasets"

PATH_ECG_EMB_TEST = os.path.join(KAGGLE_INPUT, "nadakamel3/phase2-wl/embeddings_test.npy")
PATH_GRAPH_EMB_TEST = os.path.join(KAGGLE_INPUT, "nadakamel3/embeddingsenrichis-w-tf-wl/d_graph_test.npy")
PATH_TAB_TEST = os.path.join(KAGGLE_INPUT, "nadakamel3/echonext-tabular-features/EchoNext_test_tabular_features.npy")
PATH_RAW_WAVEFORMS_TEST = os.path.join(KAGGLE_INPUT, "nadakamel3/echonext-waveforms-preprocessed/EchoNext_test_cleaned.npy")

PATH_MLP = os.path.join(KAGGLE_INPUT, "nadaelkamel3/phase-4-mlp-ecg-graph/best_mlp_phase4_ecg_graph_f2.keras")
PATH_PIPELINE_CONFIG = os.path.join(KAGGLE_INPUT, "nadaelkamel3/phase-4-mlp-ecg-graph/echonext_pipeline_config.pkl")
PATH_ATLAS_H5 = os.path.join(KAGGLE_INPUT, "nadaelkamel3/ukbrvlv/UKBRVLV.h5")
PATH_METADATA = os.path.join(KAGGLE_INPUT, "nadakamel3/echonext-metada-preprocessed/echonext_metadata_clean.csv")

TAB_FEATURE_NAMES = [
    "sex", "ventricular_rate", "atrial_rate",
    "pr_interval", "qrs_duration", "qt_corrected", "age_at_ecg",
]

LEAD_NAMES_12 = ["I", "II", "III", "aVR", "aVL", "aVF", "V1", "V2", "V3", "V4", "V5", "V6"]

SEVERITY_LABELS_FR = {
    "sain": "Sain (référence)",
    "ic_moderee": "IC modérée",
    "ic_elevee": "IC élevée",
}

PROBABILITY_THRESHOLDS = {
    "sain":       (0.00, 0.15),
    "ic_moderee": (0.15, 0.50),
    "ic_elevee":  (0.50, 1.01),
}

PHYSIO_BOUNDS = {
    0: {"min": -2.8, "max": +1.5, "soft_warn": 2.0},
    1: {"min": 0.0, "max": 2.5, "soft_warn": 1.8},
}


# ============================== 3. FICHIERS & CONFIG PIPELINE =================
def find_file(expected_path: str, search_root: str = "/kaggle/input", filename: Optional[str] = None) -> str:
    """Retourne expected_path s'il existe, sinon cherche récursivement `filename` sous search_root."""
    if os.path.exists(expected_path):
        return expected_path
    target = filename or os.path.basename(expected_path)
    for root, _, files in os.walk(search_root):
        for f in files:
            if f.lower() == target.lower():
                return os.path.join(root, f)
    raise FileNotFoundError(f"{target} introuvable sous {search_root}")


def load_pipeline_config() -> dict:
    """Charge la config du pipeline (calibrateur, seuils...), avec repli en mode dégradé."""
    try:
        pkl_path = find_file(PATH_PIPELINE_CONFIG, filename="echonext_pipeline_config.pkl")
        cfg = joblib.load(pkl_path)
        print(f"[OK] Config Pipeline chargée : {pkl_path}")
        return cfg
    except FileNotFoundError:
        warnings.warn("echonext_pipeline_config.pkl introuvable. Mode DÉGRADÉ.")
        return {
            "calibrator": None,
            "threshold_recall": 0.5,
            "tab_feature_names": TAB_FEATURE_NAMES,
        }


def load_metadata_with_id():
    """Charge les métadonnées patients et détecte la colonne d'identifiant."""
    path = find_file(PATH_METADATA, filename="echonext_metadata_clean.csv")
    df = pd.read_csv(path)
    id_col = next((c for c in ["patient_key", "eid", "patient_id", "subject_id", "id"] if c in df.columns), None)
    if id_col is None:
        df["_temp_id"] = df.index.astype(str)
        id_col = "_temp_id"
    return df, id_col


def load_metadata_aligned(n_expected: int):
    """
    Charge les métadonnées Test en réconciliant leur taille avec celle des
    tableaux d'embeddings déjà chargés (n_expected), au lieu de filtrer
    naïvement sur split=='test'. Le CSV filtré (most_recent_ecg==1) et le
    CSV complet peuvent avoir des tailles différentes pour le split test ;
    sans cette vérification, df_test.iloc[idx] peut correspondre à un
    patient différent de ecg_emb[idx]/graph_emb[idx], ce qui produit des
    probabilités affichées incohérentes pour un même "index" entre deux
    environnements (Kaggle vs Streamlit) sans que le modèle soit en cause.

    Identique à la version utilisée dans app.py (load_metadata_aligned),
    dupliquée ici pour que la Phase 5 (exécutée hors Streamlit) reste
    alignée sur les mêmes patients.
    """
    df, id_col = load_metadata_with_id()

    split_col = next((c for c in ("split", "dataset_split", "set", "fold") if c in df.columns), None)
    if split_col is None:
        raise ValueError("Colonne de split introuvable dans les métadonnées.")

    if "most_recent_ecg" in df.columns:
        df_filtre = df[df["most_recent_ecg"] == 1]
        test_filtre = df_filtre[df_filtre[split_col].astype(str).str.lower() == "test"].reset_index(drop=True)
        test_complet = df[df[split_col].astype(str).str.lower() == "test"].reset_index(drop=True)

        if len(test_filtre) == n_expected:
            return test_filtre, id_col
        elif len(test_complet) == n_expected:
            return test_complet, id_col
        else:
            raise ValueError(
                f"[DÉSALIGNEMENT METADATA / EMBEDDINGS] "
                f"embeddings = {n_expected} lignes | "
                f"CSV filtré (most_recent_ecg==1, split=test) = {len(test_filtre)} lignes | "
                f"CSV complet (split=test) = {len(test_complet)} lignes. "
                f"Impossible de garantir la correspondance patient <-> embedding — "
                f"vérifiez la source des .npy ou du CSV avant de continuer."
            )
    else:
        df_test = df[df[split_col].astype(str).str.lower() == "test"].reset_index(drop=True)
        if len(df_test) != n_expected:
            raise ValueError(
                f"[DÉSALIGNEMENT METADATA / EMBEDDINGS] "
                f"{len(df_test)} lignes de métadonnées vs {n_expected} lignes d'embeddings."
            )
        return df_test, id_col


# ============================== 4. DONNÉES & PRÉDICTION LVEF ==================
def load_ecg_derived_data():
    """Charge les embeddings ECG, graphe et les features tabulaires du split test."""
    ecg_emb = np.load(PATH_ECG_EMB_TEST)
    graph_emb = np.load(PATH_GRAPH_EMB_TEST)
    tab_scaled = np.load(PATH_TAB_TEST)
    print(f"{ecg_emb.shape[0]} patients test | ECG={ecg_emb.shape} | Graph={graph_emb.shape} | Tabulaire={tab_scaled.shape}")
    return ecg_emb, graph_emb, tab_scaled


def predict_lvef_probabilities(ecg_emb, graph_emb, tab_scaled, pipeline_config):
    """Prédit la probabilité LVEF≤45% (brute + calibrée) via le MLP multi-entrées.

    IMPORTANT : on charge le checkpoint "best F2", pas "best AUC". Le
    calibrateur isotonique (pipeline_config["calibrator"]) a été entraîné en
    Phase 4 sur les probabilités du modèle restauré par
    EarlyStopping(monitor='val_f2_score', restore_best_weights=True), donc
    sur les poids "best F2". Utiliser le checkpoint AUC ici produirait un
    mismatch calibrateur/modèle et des probabilités légèrement fausses.
    """
    mlp_path = find_file(PATH_MLP, filename="best_mlp_phase4_ecg_graph_f2.keras")
    mlp = tf.keras.models.load_model(mlp_path, compile=False, safe_mode=False)
    raw_proba = mlp.predict([ecg_emb, graph_emb], verbose=0).flatten()

    calibrator = pipeline_config.get("calibrator")
    if calibrator is not None:
        proba = calibrator.predict(raw_proba)
        proba = np.clip(proba, 0.001, 0.999)
    else:
        proba = np.clip(raw_proba, 0.001, 0.999)
        warnings.warn("Aucun calibrateur — probabilités brutes clippées.")

    return raw_proba, proba


# ============================== 5. MAILLAGE BI-VENTRICULAIRE (ATLAS UKB) ======
MESH_STRUCTURES = {
    "EPI":  dict(color="lightgray",  alpha=0.15, label="Épicarde"),
    "LV":   dict(color="firebrick",  alpha=0.95, label="Ventricule gauche (VG)"),
    "RV":   dict(color="steelblue",  alpha=0.95, label="Ventricule droit (VD)"),
    "RVFW": dict(color="royalblue",  alpha=0.55, label="Paroi libre VD"),
    "MV":   dict(color="gold",       alpha=0.95, label="Valve mitrale"),
    "AV":   dict(color="darkorange", alpha=0.95, label="Valve aortique"),
    "TV":   dict(color="khaki",      alpha=0.95, label="Valve tricuspide"),
    "PV":   dict(color="orange",     alpha=0.95, label="Valve pulmonaire"),
}

_VALVE_NAMES = {"MV", "AV", "TV", "PV"}
_CHAMBER_NAMES = {"LV", "RV", "RVFW"}
_DEFORMATION_STRUCTURES = _CHAMBER_NAMES  # structures sur lesquelles la déformation est coloriée


def mode_weights_to_score(mode_weights: dict, n_modes: Optional[int] = None) -> np.ndarray:
    """Convertit un dict {mode_idx: poids} en vecteur de score PCA dense."""
    if n_modes is None:
        n_modes = max(mode_weights.keys(), default=-1) + 1
    score = np.zeros(max(n_modes, 1))
    for idx, w in mode_weights.items():
        score[idx] = w
    return score


def get_structure_mesh(name: str, points: np.ndarray):
    """Récupère le mesh d'une structure anatomique donnée à partir des points générés par l'atlas."""
    if name == "EPI":
        return ukb_surface.get_epi_mesh(points=points)
    if name in _CHAMBER_NAMES:
        return ukb_surface.get_chamber_mesh(surface_name=name, points=points)
    if name in _VALVE_NAMES:
        return ukb_surface.get_valve_mesh(surface_name=name, points=points)
    raise ValueError(f"Structure inconnue : {name}")


@lru_cache(maxsize=128)
def _build_biv_surfaces_cached(atlas_h5_path: str, weights_key: tuple) -> Dict[str, Dict[str, object]]:
    mode_weights = dict(weights_key)
    score = mode_weights_to_score(mode_weights)
    points = ukb_atlas.generate_points(filename=atlas_h5_path, score=score)
    return {
        case_name: {name: get_structure_mesh(name, pts) for name in MESH_STRUCTURES}
        for case_name, pts in [("ED", points.ED), ("ES", points.ES)]
    }


def build_biv_surfaces(atlas_h5_path: str, mode_weights: dict) -> Dict[str, Dict[str, object]]:
    """Génère (ou récupère depuis le cache) les maillages ED/ES de toutes les structures pour
    un jeu de poids de modes PCA donné. mode_weights={} => géométrie moyenne (saine) de l'atlas.

    OPTIMISATION : la géométrie saine (mode_weights={}) et celle de chaque patient (déterminée
    uniquement par lvef_probability + qrs_duration_ms via compute_mode_weights) sont recalculées
    de nombreuses fois dans ce script — visualisation statique, chacune des 4 animations 4D,
    calcul de déformation max partagée, figure de comparaison finale... Sans cache, la même
    reconstruction PCA de l'atlas est refaite ~10-15 fois pour rien. Le cache (mémoïsation sur
    le tuple des poids) élimine tout ce travail redondant sans changer le résultat."""
    weights_key = tuple(sorted(mode_weights.items()))
    return _build_biv_surfaces_cached(atlas_h5_path, weights_key)


def compute_mode_weights(lvef_probability: float, qrs_duration_ms: float,
                          qrs_normal_upper: float = 110.0, qrs_pathological: float = 120.0,
                          max_std: float = 3.0) -> dict:
    """Traduit la proba LVEF≤45% et la durée QRS en poids sur les 2 premiers modes PCA de l'atlas."""
    w0_raw = -np.clip(lvef_probability * max_std, 0.0, max_std)
    w0 = np.clip(w0_raw, PHYSIO_BOUNDS[0]["min"], PHYSIO_BOUNDS[0]["max"])

    if qrs_duration_ms >= qrs_pathological:
        qrs_excess = qrs_duration_ms - qrs_pathological
        w1_raw = np.clip(qrs_excess / 20.0, 0.0, max_std)
    elif qrs_duration_ms >= qrs_normal_upper:
        w1_raw = 0.3 * (qrs_duration_ms - qrs_normal_upper) / (qrs_pathological - qrs_normal_upper)
    else:
        w1_raw = 0.0
    w1 = np.clip(w1_raw, PHYSIO_BOUNDS[1]["min"], PHYSIO_BOUNDS[1]["max"])

    return {0: w0, 1: w1}


def mesh_volume_ml(mesh) -> float:
    """Volume d'un maillage triangulaire fermé (formule du divergence theorem), en mL."""
    v, f = mesh.points, mesh.cells[0].data
    v0, v1, v2 = v[f[:, 0]], v[f[:, 1]], v[f[:, 2]]
    signed_vol = np.sum(np.einsum("ij,ij->i", v0, np.cross(v1, v2))) / 6.0
    return abs(signed_vol) / 1000.0


def compute_healthy_volumes(atlas_h5_path: str) -> dict:
    surfaces = build_biv_surfaces(atlas_h5_path, {})
    return {
        "ED": {"LV": mesh_volume_ml(surfaces["ED"]["LV"]), "RV": mesh_volume_ml(surfaces["ED"]["RV"])},
        "ES": {"LV": mesh_volume_ml(surfaces["ES"]["LV"]), "RV": mesh_volume_ml(surfaces["ES"]["RV"])},
    }


def report_shape_difference(surfaces_healthy: dict, surfaces_patient: dict, case: str = "ED") -> dict:
    """Écart de volume (%) VG/VD du patient par rapport à la référence saine."""
    healthy, patient = surfaces_healthy[case], surfaces_patient[case]
    return {
        name: 100 * (mesh_volume_ml(patient[name]) / mesh_volume_ml(healthy[name]) - 1)
        for name in ("LV", "RV")
    }


# ============================== 6. DÉFORMATION VS RÉFÉRENCE SAINE =============
def compute_vertex_deformation(mesh_patient, mesh_healthy) -> np.ndarray:
    """Distance euclidienne sommet-à-sommet entre le maillage patient et le maillage sain."""
    return np.linalg.norm(mesh_patient.points - mesh_healthy.points, axis=1)


def build_laplacian_smoother(faces: np.ndarray, n_vertices: int):
    """Matrice de moyennage Laplacien (voisinage 1-anneau) pour lisser un champ scalaire sur le maillage."""
    a, b, c = faces[:, 0], faces[:, 1], faces[:, 2]
    rows = np.concatenate([a, b, c, b, c, a])
    cols = np.concatenate([b, c, a, a, b, c])
    data = np.ones(len(rows))
    A = sparse.coo_matrix((data, (rows, cols)), shape=(n_vertices, n_vertices)).tocsr()
    deg = np.asarray(A.sum(axis=1)).flatten()
    deg[deg == 0] = 1.0
    return sparse.diags(1.0 / deg) @ A


def smooth_scalar_field(W, values: np.ndarray, n_iter: int = 5, lam: float = 0.6) -> np.ndarray:
    smoothed = values.astype(float).copy()
    for _ in range(n_iter):
        smoothed = (1.0 - lam) * smoothed + lam * (W @ smoothed)
    return smoothed


_LAPLACIAN_CACHE: Dict[tuple, "sparse.spmatrix"] = {}


def get_cached_laplacian(name: str, phase: str, mesh):
    """Renvoie le lisseur Laplacien pour une structure/phase donnée, en le mettant en cache.

    OPTIMISATION : la topologie (faces) d'une structure anatomique donnée, pour une phase (ED
    ou ES) donnée, est FIXE — indépendante des poids PCA du patient (seules les positions des
    sommets bougent, pas la connectivité). Ce lisseur peut donc être calculé une seule fois et
    réutilisé pour tous les patients, au lieu d'être reconstruit à chaque animation 4D et à
    chaque figure de comparaison (~4-5 fois dans ce script pour un résultat rigoureusement
    identique à chaque fois)."""
    key = (name, phase)
    if key not in _LAPLACIAN_CACHE:
        faces = mesh.cells[0].data
        _LAPLACIAN_CACHE[key] = build_laplacian_smoother(faces, len(mesh.points))
    return _LAPLACIAN_CACHE[key]


def get_4d_interpolated_mesh(surfaces_ed: dict, surfaces_es: dict, alpha: float) -> dict:
    """Interpole linéairement (ED -> ES) chaque structure pour une phase intermédiaire alpha in [0, 1]."""
    class _TempMesh:
        __slots__ = ("points", "cells")
        def __init__(self, points, cells):
            self.points = points
            self.cells = cells

    return {
        name: _TempMesh((1.0 - alpha) * surfaces_ed[name].points + alpha * surfaces_es[name].points,
                         surfaces_ed[name].cells)
        for name in MESH_STRUCTURES
    }


def compute_max_deformation_for_patient(atlas_h5_path: str, surfaces_healthy: dict, prob: float, qrs_ms: float) -> float:
    """Déformation vertex-à-vertex max (ED+ES, VG/VD/paroi libre VD) pour un patient donné."""
    weights = compute_mode_weights(prob, qrs_ms)
    surfaces_patient = build_biv_surfaces(atlas_h5_path, weights)
    local_max = 0.0
    for phase in ("ED", "ES"):
        for name in _DEFORMATION_STRUCTURES:
            d = compute_vertex_deformation(surfaces_patient[phase][name], surfaces_healthy[phase][name])
            local_max = max(local_max, float(d.max()))
    return local_max


def compute_shared_max_deformation(atlas_h5_path: str, patients_par_severite: dict,
                                    lvef_prob: np.ndarray, tab_raw: np.ndarray,
                                    margin_ratio: float = 0.05):
    """Déformation max commune à un groupe de patients, pour une échelle de couleur comparable entre eux."""
    surfaces_healthy = build_biv_surfaces(atlas_h5_path, {})
    global_max, details = 0.0, {}
    for label, info in patients_par_severite.items():
        idx = info["index"]
        prob = float(lvef_prob[idx])
        qrs_ms = float(tab_raw[idx, TAB_FEATURE_NAMES.index("qrs_duration")])
        local_max = compute_max_deformation_for_patient(atlas_h5_path, surfaces_healthy, prob, qrs_ms)
        details[label] = local_max
        global_max = max(global_max, local_max)
    return global_max * (1.0 + margin_ratio), details


# ============================== 7. RENDU 3D UNIFIÉ : MAILLAGE + CONTOUR BLANC =
# La seule et unique fonction de rendu : elle trace la surface colorée du
# patient (avec overlay de déformation optionnel) ET, si une référence saine
# est fournie, superpose son contour filaire blanc — ce qui remplace les
# anciennes fonctions dupliquées plot_biv_mesh / plot_biv_mesh_4d /
# add_reference_wireframe.

def shade_triangles(tris: np.ndarray, base_color: str, alpha: float,
                     light_direction: Iterable[float] = (-1.0, -1.0, 1.2),
                     ambient: float = 0.35, diffuse: float = 0.65) -> np.ndarray:
    """Couleur RGBA par triangle selon un éclairage directionnel simple (shading Lambertien)."""
    base = np.array(to_rgb(base_color))
    v0, v1, v2 = tris[:, 0], tris[:, 1], tris[:, 2]

    normals = np.cross(v1 - v0, v2 - v0)
    norm_len = np.linalg.norm(normals, axis=1, keepdims=True)
    norm_len[norm_len == 0] = 1.0
    normals /= norm_len

    light = np.asarray(light_direction, dtype=float)
    light /= np.linalg.norm(light)

    intensity = np.clip(np.abs(normals @ light), 0, 1)
    shade = ambient + diffuse * intensity

    rgb = np.clip(base[None, :] * shade[:, None], 0, 1)
    return np.concatenate([rgb, np.full((rgb.shape[0], 1), alpha)], axis=1)


def _mesh_triangles(mesh):
    """Retourne (triangles (N,3,3), faces (N,3)) d'un mesh, ou None si invalide."""
    try:
        faces = np.asarray(mesh.cells[0].data)[:, :3].astype(int)
    except Exception:
        return None
    return np.asarray(mesh.points)[faces], faces


def _mesh_wireframe_segments(mesh) -> Optional[np.ndarray]:
    """Retourne les segments d'arêtes uniques (N,2,3) d'un mesh, ou None si invalide."""
    result = _mesh_triangles(mesh)
    if result is None:
        return None
    points, faces = np.asarray(mesh.points), result[1]
    edges = np.concatenate([faces[:, [0, 1]], faces[:, [1, 2]], faces[:, [2, 0]]], axis=0)
    edges = np.unique(np.sort(edges, axis=1), axis=0)
    return points[edges]


def render_biv_mesh(
    ax,
    surfaces_patient: Dict[str, object],
    *,
    # --- overlay de déformation (optionnel) ---
    deformation_by_structure: Optional[Dict[str, np.ndarray]] = None,
    blend_weight: float = 0.6,
    max_deform: float = 8.0,
    cmap_name: str = "inferno",
    # --- référence saine : contour filaire blanc (optionnel) ---
    surfaces_reference: Optional[Dict[str, object]] = None,
    reference_structures: Optional[Iterable[str]] = None,
    reference_color: str = "white",
    reference_linewidth: float = 0.7,
    reference_alpha: float = 0.45,
    show_reference_label: bool = True,
    # --- mise en page ---
    title: str = "",
    title_color: str = "white",
    view_bounds: Optional[tuple] = None,
    # --- perf : sous-ensemble de structures à dessiner (None = toutes) ---
    render_structures: Optional[Iterable[str]] = None,
) -> None:
    """
    Trace en un seul appel :
      - la surface 3D colorée du patient (avec overlay de déformation optionnel) ;
      - si `surfaces_reference` est fourni, le contour filaire blanc de la
        géométrie saine, superposé pour visualiser directement l'écart
        patient / référence.

    C'est la fonction unique qui remplace les anciennes plot_biv_mesh,
    plot_biv_mesh_4d (dupliquée) et add_reference_wireframe.
    """
    ax.clear()

    # --- 1) Surface colorée du patient (+ overlay de déformation) ---
    norm = Normalize(vmin=0.0, vmax=max_deform)
    cmap = matplotlib.colormaps[cmap_name]  # remplace cm.get_cmap(), déprécié depuis MPL 3.7
    all_points = []

    structures_to_draw = set(render_structures) if render_structures is not None else None

    for name, style in MESH_STRUCTURES.items():
        if structures_to_draw is not None and name not in structures_to_draw:
            continue
        mesh = surfaces_patient.get(name)
        if mesh is None:
            continue
        tris_faces = _mesh_triangles(mesh)
        if tris_faces is None:
            continue
        tris, faces = tris_faces

        rgba = shade_triangles(tris, style["color"], style["alpha"])

        has_deform = (
            deformation_by_structure is not None
            and name in _DEFORMATION_STRUCTURES
            and name in deformation_by_structure
        )
        if has_deform:
            tri_deform = deformation_by_structure[name][faces].mean(axis=1)
            overlay = cmap(norm(tri_deform))
            overlay[:, 3] = rgba[:, 3]
            rgba = (1.0 - blend_weight) * rgba + blend_weight * overlay

        ax.add_collection3d(Poly3DCollection(tris, facecolor=rgba, edgecolor="none"))
        all_points.append(mesh.points)

    # --- 2) Contour filaire blanc de la géométrie saine (référence) ---
    if surfaces_reference is not None:
        structures = list(reference_structures) if reference_structures else list(surfaces_reference)
        segments = []
        for name in structures:
            mesh = surfaces_reference.get(name)
            if mesh is None:
                continue
            seg = _mesh_wireframe_segments(mesh)
            if seg is not None:
                segments.append(seg)

        if segments:
            wire = Line3DCollection(
                np.concatenate(segments, axis=0),
                colors=reference_color,
                linewidths=reference_linewidth,
                alpha=reference_alpha,
                linestyles="solid",
            )
            ax.add_collection3d(wire)

            if show_reference_label:
                ax.text2D(
                    0.03, 0.04, "Contour blanc = référence saine",
                    transform=ax.transAxes, color="white", fontsize=8, alpha=0.75,
                )

    # --- 3) Cadrage, titre, habillage ---
    if view_bounds is not None:
        (xmin, xmax), (ymin, ymax), (zmin, zmax) = view_bounds
    elif all_points:
        pts = np.concatenate(all_points, axis=0)
        mins, maxs = pts.min(axis=0), pts.max(axis=0)
        center = (mins + maxs) / 2.0
        half = (maxs - mins).max() / 2.0 * 1.12
        (xmin, xmax), (ymin, ymax), (zmin, zmax) = [(c - half, c + half) for c in center]
    else:
        (xmin, xmax), (ymin, ymax), (zmin, zmax) = (-60, 60), (-60, 60), (-60, 60)

    ax.set_xlim(xmin, xmax)
    ax.set_ylim(ymin, ymax)
    ax.set_zlim(zmin, zmax)
    ax.set_box_aspect([1, 1, 1])
    ax.set_title(title, fontsize=11, fontweight="bold", pad=5, color=title_color)
    ax.set_axis_off()


def compute_shared_view_bounds(surface_sets: Iterable[Dict[str, object]], padding_ratio: float = 0.12) -> tuple:
    """Cube englobant commun (mêmes limites x/y/z) pour plusieurs jeux de maillages."""
    all_pts = [mesh.points for surfaces in surface_sets for mesh in surfaces.values()]
    all_pts = np.concatenate(all_pts, axis=0)
    mins, maxs = all_pts.min(axis=0), all_pts.max(axis=0)
    center = (mins + maxs) / 2.0
    half_range = (maxs - mins).max() / 2.0 * (1.0 + padding_ratio)
    return tuple((c - half_range, c + half_range) for c in center)


def add_structure_legend(target, ncol: int = 4, text_color: str = "white", loc: str = "center"):
    """Légende commune des structures anatomiques ; `target` peut être un fig ou un ax."""
    handles = [Patch(facecolor=s["color"], alpha=s["alpha"], label=s["label"]) for s in MESH_STRUCTURES.values()]
    return target.legend(handles=handles, loc=loc, ncol=ncol, frameon=False, fontsize=9, labelcolor=text_color)


# ============================== 8. ECG (contexte du jumeau numérique) =========
def extract_real_ecg_beat(raw_waveforms, patient_idx: int, n_frames: int = 30,
                           lead_idx: int = 1, hr_bpm: float = 60.0, fs: int = 250):
    """Extrait et ré-échantillonne un battement ECG réel (dérivation `lead_idx`) sur n_frames points."""
    signal = np.asarray(raw_waveforms[patient_idx, :, lead_idx], dtype=float)
    n_samples = len(signal)
    rr_samples = max(int(round((60.0 / max(hr_bpm, 1e-3)) * fs)), 10)

    center = n_samples // 2
    search_lo, search_hi = max(0, center - rr_samples), min(n_samples, center + rr_samples)
    local_r_idx = search_lo + int(np.argmax(np.abs(signal[search_lo:search_hi])))

    half_rr = rr_samples // 2
    start, end = max(0, local_r_idx - half_rr), min(n_samples, local_r_idx + half_rr)
    beat = signal[start:end]
    if len(beat) < 5:
        beat = signal[:rr_samples] if len(signal) >= rr_samples else signal

    t_original = np.linspace(0, 1, len(beat))
    t_ecg = np.linspace(0, 1, n_frames)
    beat_resampled = np.interp(t_ecg, t_original, beat)

    b_min, b_max = beat_resampled.min(), beat_resampled.max()
    beat_norm = (
        -0.3 + 1.3 * (beat_resampled - b_min) / (b_max - b_min)
        if b_max - b_min > 1e-9 else np.zeros_like(beat_resampled)
    )
    return t_ecg, beat_norm


def plot_full_ecg_verification(raw_waveforms, df_test, id_col, tab_raw, patient_idx, fs=250, save_path=None):
    """Vérifie la cohérence FC déclarée / FC estimée sur le signal réel, et affiche les 12 dérivations."""
    signal_12 = np.asarray(raw_waveforms[patient_idx], dtype=float)
    n_samples = signal_12.shape[0]
    t = np.arange(n_samples) / fs
    duration_s = n_samples / fs

    patient_id = df_test.iloc[patient_idx][id_col] if id_col in df_test.columns else f"Idx_{patient_idx}"
    hr_declared = float(tab_raw[patient_idx, TAB_FEATURE_NAMES.index("ventricular_rate")])
    qrs_declared = float(tab_raw[patient_idx, TAB_FEATURE_NAMES.index("qrs_duration")])

    lead_ii = signal_12[:, 1]
    peak_height = lead_ii.mean() + 2.0 * lead_ii.std()
    peaks, _ = find_peaks(lead_ii, height=peak_height, distance=int(0.3 * fs))
    n_beats = len(peaks)
    hr_estimated = (n_beats / duration_s) * 60.0 if duration_s > 0 and n_beats >= 2 else float("nan")
    ecart = abs(hr_estimated - hr_declared) if not np.isnan(hr_estimated) else float("nan")

    print(f"[VÉRIFICATION ECG] Patient {patient_id} (index #{patient_idx})")
    print(f"  • FC déclarée (données tabulaires)  : {hr_declared:.1f} BPM")
    print(f"  • FC estimée depuis le SIGNAL RÉEL  : {hr_estimated:.1f} BPM ({n_beats} pics R / {duration_s:.1f}s)")
    print(f"  • Durée QRS déclarée (tabulaire)    : {qrs_declared:.1f} ms")
    if n_beats < 2:
        print("  ⚠️ Détection de pics R insuffisante — vérifier visuellement le tracé ci-dessous.")
    elif ecart <= 15:
        print(f"  ✅ FC signal réel ≈ FC tabulaire (écart {ecart:.1f} BPM).")
    else:
        print(f"  ⚠️ Écart FC important ({ecart:.1f} BPM) — vérifier l'alignement raw_waveforms_test <-> df_test.")

    fig, axes = plt.subplots(12, 1, figsize=(12, 13), sharex=True, facecolor='#0d1117')
    fig.suptitle(f"ECG complet (10 s, 12 dérivations) — Patient {patient_id}",
                 color='white', fontsize=13, fontweight='bold', y=0.995)
    for lead_i, ax in enumerate(axes):
        ax.set_facecolor('#0d1117')
        ax.plot(t, signal_12[:, lead_i], color='#00ffcc', lw=0.8)
        for p in peaks:
            ax.axvline(t[p], color='#ff3366', lw=0.6, linestyle='--', alpha=0.6)
        ax.set_ylabel(LEAD_NAMES_12[lead_i], color='white', fontsize=9, rotation=0, labelpad=22, va='center')
        ax.set_yticks([])
        ax.tick_params(colors='white', labelsize=7)
        for spine in ax.spines.values():
            spine.set_color('#30363d')
    axes[-1].set_xlabel("Temps (s)", color='white')
    axes[-1].tick_params(colors='white')
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=100, facecolor=fig.get_facecolor())
        print(f"[OK] Figure sauvegardée : {save_path}")
    plt.show()


# ============================== 9. VISUALISATION 3D (patient unique) ==========
def visualize_patient(atlas_h5_path, ecg_emb, graph_emb, tab_scaled, tab_raw,
                       lvef_probabilities, healthy_vols, patient_idx=None, title_suffix=""):
    """
    Affiche côte à côte, en 3D statique :
      - la référence saine seule ;
      - le patient AVEC le contour blanc de la référence saine en superposition
        (pour visualiser directement l'écart géométrique).
    """
    idx = patient_idx if patient_idx is not None else 0
    prob = float(lvef_probabilities[idx])
    qrs = float(tab_raw[idx, TAB_FEATURE_NAMES.index("qrs_duration")])
    hr = float(tab_raw[idx, TAB_FEATURE_NAMES.index("ventricular_rate")])

    surfaces_healthy = build_biv_surfaces(atlas_h5_path, {})
    weights = compute_mode_weights(prob, qrs)
    surfaces_patient = build_biv_surfaces(atlas_h5_path, weights)
    pct_change = report_shape_difference(surfaces_healthy, surfaces_patient, case="ED")

    fig = plt.figure(figsize=(12, 6))

    ax1 = fig.add_subplot(121, projection="3d")
    render_biv_mesh(ax1, surfaces_healthy["ED"], title="Référence saine (ED)", title_color="black")

    ax2 = fig.add_subplot(122, projection="3d")
    render_biv_mesh(
        ax2, surfaces_patient["ED"],
        title=f"Patient #{idx}{title_suffix}\nProba LVEF≤45% = {prob:.2f}", title_color="black",
        surfaces_reference=surfaces_healthy["ED"], reference_structures=_DEFORMATION_STRUCTURES,
    )

    add_structure_legend(fig, loc="lower center")
    plt.tight_layout()
    plt.show()

    return {
        "patient_idx": int(idx),
        "lvef_probability": prob,
        "qrs_duration_ms": qrs,
        "heart_rate_bpm": hr,
        "mode_weights": weights,
        "volume_pct_change": pct_change,
    }


# ============================== 10. JUMEAU NUMÉRIQUE 4D (animation) ===========
def create_dynamic_digital_twin(atlas_h5_path, patient_idx, patient_id_label, lvef_probabilities, tab_raw,
                                 raw_waveforms, lead_idx=1, n_frames=20, save_path="jumeau_numerique_4D.gif",
                                 show_deformation=True, max_deform=8.0, blend_weight=0.6, cmap_name="inferno",
                                 smoothing_iterations=5, smoothing_lambda=0.6, fast_mode=False, dpi=85):
    """Génère l'animation 4D (cycle cardiaque complet) : panneau sain vs panneau patient (coloré par déformation).

    PERF :
      - n_frames=20 (au lieu de 30) et dpi=85 (au lieu de 100) par défaut : le rendu 3D par
        frame (tri en profondeur + rastérisation Agg des Poly3DCollection) est le vrai coût,
        pas les calculs numpy — moins de pixels et de frames à produire = génération plus
        rapide, sans perte de fluidité perceptible sur un GIF.
      - fast_mode=True : n'affiche que VG/VD/paroi libre VD (celles qui portent la couleur de
        déformation) et saute l'épicarde + les 4 valves, qui sont surtout décoratifs pendant
        l'animation. Coupe le nombre de Poly3DCollection créées/triées par frame d'environ
        moitié → gain net sur le temps de génération. À utiliser pour les brouillons ; repasser
        à False pour la version finale/soutenance si tu veux l'épicarde + les valves visibles.
    """
    print(f"[INFO] Génération du Jumeau Numérique 4D pour le Patient {patient_id_label} (Index #{patient_idx})...")

    prob_lvef = float(lvef_probabilities[patient_idx])
    qrs_ms = float(tab_raw[patient_idx, TAB_FEATURE_NAMES.index("qrs_duration")])
    hr_bpm = float(tab_raw[patient_idx, TAB_FEATURE_NAMES.index("ventricular_rate")])

    surfaces_healthy = build_biv_surfaces(atlas_h5_path, {})
    weights_patient = compute_mode_weights(prob_lvef, qrs_ms)
    surfaces_patient = build_biv_surfaces(atlas_h5_path, weights_patient)

    view_bounds = compute_shared_view_bounds([
        surfaces_healthy["ED"], surfaces_healthy["ES"], surfaces_patient["ED"], surfaces_patient["ES"],
    ])

    t_ecg, ecg_signal = extract_real_ecg_beat(raw_waveforms, patient_idx, n_frames=n_frames,
                                               lead_idx=lead_idx, hr_bpm=hr_bpm)

    deformation_smooth = {"ED": {}, "ES": {}}
    if show_deformation:
        for phase in ("ED", "ES"):
            for name in _DEFORMATION_STRUCTURES:
                mesh_pat, mesh_hea = surfaces_patient[phase][name], surfaces_healthy[phase][name]
                raw_deform = compute_vertex_deformation(mesh_pat, mesh_hea)
                W = get_cached_laplacian(name, phase, mesh_pat)
                deformation_smooth[phase][name] = smooth_scalar_field(
                    W, raw_deform, n_iter=smoothing_iterations, lam=smoothing_lambda
                )

    fig = plt.figure(figsize=(14, 11.5), facecolor='#0d1117')
    gs = GridSpec(5, 2, height_ratios=[4.2, 0.55, 0.55, 0.85, 0.95], figure=fig, hspace=0.5, wspace=0.02)

    ax_healthy = fig.add_subplot(gs[0, 0], projection='3d', facecolor='#0d1117')
    ax_patient = fig.add_subplot(gs[0, 1], projection='3d', facecolor='#0d1117')
    ax_colorbar = fig.add_subplot(gs[1, :], facecolor='#0d1117'); ax_colorbar.axis('off')
    ax_legend = fig.add_subplot(gs[2, :], facecolor='#0d1117'); ax_legend.axis('off')
    ax_physio = fig.add_subplot(gs[3, :], facecolor='#0d1117'); ax_physio.axis('off')
    ax_ecg = fig.add_subplot(gs[4, :], facecolor='#161b22')

    add_structure_legend(ax_legend, ncol=4, text_color='white')

    if show_deformation:
        ax_colorbar.text(0.5, 0.92, "Que signifient les couleurs ?", ha='center', va='top',
                          color='white', fontsize=10, fontweight='bold', transform=ax_colorbar.transAxes)
        ax_colorbar.text(0.5, 0.68,
                          "Écart de paroi entre le cœur du patient et la référence saine, au même point anatomique (en mm)",
                          ha='center', va='top', color='#c9d1d9', fontsize=8.5, transform=ax_colorbar.transAxes)
        cax = ax_colorbar.inset_axes([0.20, 0.02, 0.60, 0.28])
        sm = cm.ScalarMappable(norm=Normalize(vmin=0.0, vmax=max_deform), cmap=cmap_name)
        sm.set_array([])
        cb = fig.colorbar(sm, cax=cax, orientation='horizontal')
        cb.ax.xaxis.set_tick_params(color='white', labelsize=8)
        plt.setp(plt.getp(cb.ax.axes, 'xticklabels'), color='white')
        ax_colorbar.text(0.19, 0.30, "0 mm\n(identique au cœur sain)", ha='right', va='bottom',
                          color='#9aa4ad', fontsize=7.5, transform=ax_colorbar.transAxes)
        ax_colorbar.text(0.81, 0.30, f"{max_deform:.0f} mm\n(déformation la plus marquée)", ha='left', va='bottom',
                          color='#9aa4ad', fontsize=7.5, transform=ax_colorbar.transAxes)

    ax_ecg.plot(t_ecg, ecg_signal, color='#00ffcc', lw=2, label="Signal ECG réel (dérivation II)")
    ecg_cursor, = ax_ecg.plot([0, 0], [-0.5, 1.2], color='#ff3366', lw=2, linestyle='--')
    ax_ecg.set_ylim(-0.5, 1.3); ax_ecg.set_xlim(0, 1); ax_ecg.axis('off')

    title_text = fig.suptitle("", fontsize=14, fontweight='bold', color='white', y=0.97)
    physio_text = ax_physio.text(0.02, 0.5, "", transform=ax_physio.transAxes, va='center',
                                  color='white', fontsize=11, fontfamily='monospace',
                                  bbox=dict(boxstyle='round', facecolor='#21262d', edgecolor='#30363d', alpha=0.8))

    animation_structures = _DEFORMATION_STRUCTURES if fast_mode else None

    def update(frame):
        cycle_progress = frame / n_frames
        alpha = 0.5 * (1.0 - np.cos(cycle_progress * 2 * np.pi))

        mesh_hea_4d = get_4d_interpolated_mesh(surfaces_healthy["ED"], surfaces_healthy["ES"], alpha)
        mesh_pat_4d = get_4d_interpolated_mesh(surfaces_patient["ED"], surfaces_patient["ES"], alpha)

        deform_frame = None
        if show_deformation:
            deform_frame = {
                name: (1.0 - alpha) * deformation_smooth["ED"][name] + alpha * deformation_smooth["ES"][name]
                for name in _DEFORMATION_STRUCTURES
            }

        # Panneau gauche : référence saine seule.
        render_biv_mesh(ax_healthy, mesh_hea_4d, title="RÉFÉRENCE SAINE (Normo-kinésie)",
                         view_bounds=view_bounds, render_structures=animation_structures)

        # Panneau droit : patient coloré par déformation, AVEC contour blanc de la référence saine.
        render_biv_mesh(
            ax_patient, mesh_pat_4d,
            deformation_by_structure=deform_frame, blend_weight=blend_weight,
            max_deform=max_deform, cmap_name=cmap_name,
            title=f"JUMEAU PATIENT ({patient_id_label})", view_bounds=view_bounds,
            surfaces_reference=mesh_hea_4d, reference_structures=_DEFORMATION_STRUCTURES,
            reference_linewidth=0.75, reference_alpha=0.50,
            render_structures=animation_structures,
        )

        ecg_cursor.set_xdata([t_ecg[frame], t_ecg[frame]])

        vol_lv_pat = mesh_volume_ml(mesh_pat_4d["LV"])
        vol_rv_pat = mesh_volume_ml(mesh_pat_4d["RV"])
        phase_name = "SYSTOLE (Contraction)" if 0.25 <= cycle_progress <= 0.75 else "DIASTOLE (Remplissage)"

        title_text.set_text(f"Jumeau Numérique 4D — Patient {patient_id_label} | Phase : {phase_name}")
        physio_text.set_text(
            f" [PARAMÈTRES PHYSIOLOGIQUES EN TEMPS RÉEL]\n"
            f" ├─ Fréquence Cardiaque : {hr_bpm:.0f} BPM\n"
            f" ├─ Durée QRS           : {qrs_ms:.1f} ms\n"
            f" ├─ Proba LVEF ≤ 45%    : {prob_lvef*100:.1f} %\n"
            f" └─ Vol. Instantané VG : {vol_lv_pat:.1f} mL | VD : {vol_rv_pat:.1f} mL"
        )
        return ax_healthy, ax_patient, ecg_cursor, physio_text

    ani = animation.FuncAnimation(fig, update, frames=n_frames, interval=80, blit=False)
    writer = animation.FFMpegWriter(fps=12, bitrate=1800) if save_path.endswith('.mp4') else animation.PillowWriter(fps=12)
    ani.save(save_path, writer=writer, dpi=dpi)
    plt.close(fig)
    print(f"[SUCCÈS] Animation 4D sauvegardée sous : {save_path}")

    if save_path.endswith('.gif'):
        from IPython.display import Image, display
        display(Image(open(save_path, 'rb').read()))


# ============================== 11. SÉLECTION DE PATIENTS PAR SÉVÉRITÉ ========
def select_patients_by_probability_thresholds(lvef_prob: np.ndarray, thresholds: Optional[dict] = None) -> dict:
    """Sélectionne un patient représentatif par tranche de probabilité LVEF≤45%."""
    thresholds = thresholds or PROBABILITY_THRESHOLDS
    selection, used_indices = {}, set()

    for label, (low, high) in thresholds.items():
        center = (low + high) / 2.0
        mask = (lvef_prob >= low) & (lvef_prob < high)
        candidates = np.array([i for i in np.where(mask)[0] if i not in used_indices])

        if candidates.size == 0:
            warnings.warn(f"Aucun patient trouvé dans la tranche '{label}' {low}-{high} ; "
                           f"repli sur le patient le plus proche du centre de la tranche.")
            idx = int(np.argmin(np.abs(lvef_prob - center)))
        else:
            sub_probs = lvef_prob[candidates]
            idx = int(candidates[np.argmin(np.abs(sub_probs - center))])

        used_indices.add(idx)
        selection[label] = {"index": idx, "lvef_probability": float(lvef_prob[idx]), "seuils": (low, high)}

    return selection


# ============================== 12. FIGURE DE COMPARAISON 3 PATIENTS ==========
def plot_comparisonvf_3_patients(
    atlas_h5_path, patients_par_severite, lvef_prob, tab_raw, df_test, id_col,
    shared_max_deform, max_deform_par_patient, phase="ED", cmap_name="inferno",
    save_path="/kaggle/working/comparaison_3_patients.png",
):
    """
    Compare 3 patients côte à côte, avec pour chacun :
      - surface colorée = géométrie du patient (couleur = écart vs référence) ;
      - contour blanc   = géométrie saine de référence, superposée ;
      - même échelle de couleur (shared_max_deform) pour une comparaison honnête.
    """
    order = ["sain", "ic_moderee", "ic_elevee"]
    surfaces_healthy = build_biv_surfaces(atlas_h5_path, {})

    fig = plt.figure(figsize=(18, 8.5), facecolor='#0d1117')
    gs = GridSpec(1, 3, figure=fig, top=0.80, bottom=0.16, wspace=0.05)

    axes_3d, deform_texts = [], []

    for col, label in enumerate(order):
        info = patients_par_severite[label]
        idx = info["index"]
        pid = df_test.iloc[idx][id_col] if id_col in df_test.columns else f"Idx_{idx}"
        prob = float(lvef_prob[idx])
        qrs_ms = float(tab_raw[idx, TAB_FEATURE_NAMES.index("qrs_duration")])

        weights = compute_mode_weights(prob, qrs_ms)
        surfaces_patient = build_biv_surfaces(atlas_h5_path, weights)

        deform_by_structure = {}
        for name in _DEFORMATION_STRUCTURES:
            mesh_pat, mesh_hea = surfaces_patient[phase][name], surfaces_healthy[phase][name]
            raw_deform = compute_vertex_deformation(mesh_pat, mesh_hea)
            W = get_cached_laplacian(name, phase, mesh_pat)
            deform_by_structure[name] = smooth_scalar_field(W, raw_deform, n_iter=5, lam=0.6)

        ax = fig.add_subplot(gs[0, col], projection='3d', facecolor='#0d1117')
        title = f"{SEVERITY_LABELS_FR[label]}\nPatient {pid} | Proba LVEF≤45% = {prob*100:.1f}%"

        # --- Un seul appel : surface patient (déformation) + contour blanc de référence ---
        render_biv_mesh(
            ax, surfaces_patient[phase],
            deformation_by_structure=deform_by_structure, blend_weight=0.6,
            max_deform=shared_max_deform, cmap_name=cmap_name, title=title,
            surfaces_reference=surfaces_healthy[phase], reference_structures=_DEFORMATION_STRUCTURES,
            reference_color="white", reference_linewidth=0.75, reference_alpha=0.50,
        )

        axes_3d.append(ax)
        x_center = (col + 0.5) / 3.0
        deform_texts.append((x_center, label))

    for x_center, label in deform_texts:
        fig.text(
            x_center, 0.06,
            f"Écart géométrique max : {max_deform_par_patient[label]:.1f} mm\n"
            f"Échelle couleur : 0–{shared_max_deform:.0f} mm",
            ha='center', va='center', color='#e6edf3', fontsize=10.5, fontfamily='monospace', zorder=100,
            bbox=dict(boxstyle='round,pad=0.4', facecolor='#21262d', edgecolor='#30363d', alpha=0.95),
        )

    # Même cadrage spatial pour les 3 patients.
    view_bounds = compute_shared_view_bounds([
        surfaces_healthy[phase],
        *[build_biv_surfaces(atlas_h5_path, compute_mode_weights(
            float(lvef_prob[patients_par_severite[l]["index"]]),
            float(tab_raw[patients_par_severite[l]["index"], TAB_FEATURE_NAMES.index("qrs_duration")]),
        ))[phase] for l in order],
    ])
    (xmin, xmax), (ymin, ymax), (zmin, zmax) = view_bounds
    for ax in axes_3d:
        ax.set_xlim(xmin, xmax); ax.set_ylim(ymin, ymax); ax.set_zlim(zmin, zmax)
        ax.set_box_aspect([1, 1, 1])

    phase_label = "Diastole (ED)" if phase == "ED" else "Systole (ES)"
    fig.suptitle(f"Comparaison des 3 patients — {phase_label}", fontsize=15, fontweight='bold', color='white', y=1.02)

    cax = fig.add_axes([0.30, 0.965, 0.40, 0.02])
    sm = cm.ScalarMappable(norm=Normalize(vmin=0.0, vmax=shared_max_deform), cmap=cmap_name)
    sm.set_array([])
    cb = fig.colorbar(sm, cax=cax, orientation='horizontal')
    cb.set_label("Écart géométrique vs référence saine (mm)", color='white', fontsize=9)
    cb.ax.xaxis.set_tick_params(color='white', labelsize=8)
    plt.setp(plt.getp(cb.ax.axes, 'xticklabels'), color='white')

    fig.text(
        0.5, 0.015,
        "Surface colorée = géométrie du patient    |    Contour blanc = géométrie saine de référence",
        ha='center', va='center', color='#e6edf3', fontsize=9,
    )

    plt.savefig(save_path, dpi=150, facecolor=fig.get_facecolor(), bbox_inches='tight')
    plt.show()
    print(f"[OK] Figure de comparaison sauvegardée : {save_path}")
    return save_path


# ============================== 13. ÉTUDE D'ABLATION (RÉINTÉGRÉE) =============
# Compare l'apport de chaque modalité sur la performance de détection :
#   1. ECG Embeddings bruts seul
#   2. ECG Embeddings + Données Tabulaires
#   3. Modèle Complet (ECG + Graph Enrichi + Tabulaire) — celui de la Phase 4
#
# Correction conservée par rapport à la version d'origine du notebook :
# split Train/Test strict (train_test_split, stratifié) pour l'entraînement à
# la volée des deux baselines, afin d'éviter toute fuite de données (data
# leakage) entre l'entraînement et l'évaluation.
# Bug corrigé par rapport au code fourni : `pred_ecg_tab` se retrouvait
# transformé en tuple à cause d'une virgule flottante en fin de ligne
# (`.flatten(),`) — supprimée ici.

def build_baseline_mlp_ecg_only(input_dim_ecg):
    """Modèle d'Ablation 1 : ECG Embeddings bruts seul."""
    inp = layers.Input(shape=(input_dim_ecg,), name="ecg_input")
    x = layers.Dense(128, activation="relu")(inp)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(0.2)(x)
    out = layers.Dense(1, activation="sigmoid", name="output")(x)

    model = models.Model(inputs=inp, outputs=out, name="MLP_ECG_Only")
    model.compile(optimizer=optimizers.Adam(1e-3), loss="binary_crossentropy", metrics=["AUC"])
    return model


def build_baseline_mlp_ecg_tab(input_dim_ecg, input_dim_tab):
    """Modèle d'Ablation 2 : ECG Embeddings + Données Tabulaires."""
    inp_ecg = layers.Input(shape=(input_dim_ecg,), name="ecg_input")
    inp_tab = layers.Input(shape=(input_dim_tab,), name="tab_input")

    x1 = layers.Dense(128, activation="relu")(inp_ecg)
    x2 = layers.Dense(32, activation="relu")(inp_tab)

    concat = layers.Concatenate()([x1, x2])
    x = layers.Dense(64, activation="relu")(concat)
    x = layers.Dropout(0.2)(x)
    out = layers.Dense(1, activation="sigmoid", name="output")(x)

    model = models.Model(inputs=[inp_ecg, inp_tab], outputs=out, name="MLP_ECG_Tabular")
    model.compile(optimizer=optimizers.Adam(1e-3), loss="binary_crossentropy", metrics=["AUC"])
    return model


# ============================== 13. ÉTUDE D'ABLATION (CORRIGÉE) ===============
# Chaîne de fusion réelle du pipeline :
#   Phase 2 : waveforms + 7 features tabulaires -> embedding ECG "brut" (128D).
#             Le tabulaire est fusionné UNE SEULE FOIS ici, implicitement.
#   Phase 3 : embedding ECG (déjà tab-infusé) -> embedding "enrichi" par le
#             graphe. Aucune réinjection explicite du tabulaire à ce stade.
#   Phase 4 : MLP final = ECG embedding + Graph embedding UNIQUEMENT.
#             Toujours aucune entrée tabulaire explicite (2 entrées, pas 3).
#
# Cette étude compare donc :
#   1. ECG (tab fusionné implicitement en Phase 2, jamais réinjecté après)
#   2. ECG + Tabulaire réinjecté explicitement — baseline comparative créée
#      UNIQUEMENT pour cette ablation (n'existe nulle part ailleurs dans le
#      pipeline réel), pour mesurer si exposer le MLP directement au signal
#      tabulaire déjà présent (mais diffus) dans l'embedding apporte un gain.
#   3. ECG + Graph (Phase 3/4), sans réinjection tabulaire — c'est le modèle
#      final réellement utilisé en production.
#
# Split Train/Test strict pour l'entraînement à la volée de la baseline 2,
# afin d'éviter tout data leakage.

def build_baseline_mlp_ecg_only(input_dim_ecg):
    """Config 1 : Embedding ECG seul (le tabulaire y est déjà fusionné
    implicitement depuis la Phase 2, mais jamais réinjecté explicitement)."""
    inp = layers.Input(shape=(input_dim_ecg,), name="ecg_input")
    x = layers.Dense(128, activation="relu")(inp)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(0.2)(x)
    out = layers.Dense(1, activation="sigmoid", name="output")(x)

    model = models.Model(inputs=inp, outputs=out, name="MLP_ECG_Only")
    model.compile(optimizer=optimizers.Adam(1e-3), loss="binary_crossentropy", metrics=["AUC"])
    return model


def build_baseline_mlp_ecg_tab(input_dim_ecg, input_dim_tab):
    """Config 2 : Embedding ECG + réinjection EXPLICITE des 7 features
    tabulaires. Baseline comparative ad hoc, créée uniquement pour cette
    étude d'ablation — ne fait pas partie du pipeline Phase 2/3/4 réel."""
    inp_ecg = layers.Input(shape=(input_dim_ecg,), name="ecg_input")
    inp_tab = layers.Input(shape=(input_dim_tab,), name="tab_input")

    x1 = layers.Dense(128, activation="relu")(inp_ecg)
    x2 = layers.Dense(32, activation="relu")(inp_tab)

    concat = layers.Concatenate()([x1, x2])
    x = layers.Dense(64, activation="relu")(concat)
    x = layers.Dropout(0.2)(x)
    out = layers.Dense(1, activation="sigmoid", name="output")(x)

    model = models.Model(inputs=[inp_ecg, inp_tab], outputs=out, name="MLP_ECG_Tabular")
    model.compile(optimizer=optimizers.Adam(1e-3), loss="binary_crossentropy", metrics=["AUC"])
    return model


def run_ablation_study(pipeline_config):
    """
    Exécute l'étude d'ablation comparant :
      1. ECG seul (tab déjà fusionné implicitement en Phase 2)
      2. ECG + Tabulaire réinjecté explicitement (baseline comparative Phase 5)
      3. ECG + Graph (modèle final réel, Phase 4 — 2 entrées, sans tabulaire explicite)

    Split Train/Test strict pour l'entraînement à la volée de la baseline 2,
    afin d'éviter tout data leakage.
    """
    print("\n" + "=" * 60)
    print(" LANCEMENT DE L'ÉTUDE D'ABLATION (APPORT DU GRAPH ENRICHMENT")
    print(" VS RÉINJECTION EXPLICITE DU TABULAIRE)")
    print("=" * 60)

    ecg_emb, graph_emb, tab_scaled = load_ecg_derived_data()

    # Même fix d'alignement que dans main() : on réconcilie df_test avec la
    # longueur réelle des embeddings AVANT d'en extraire les labels, plutôt
    # que de tronquer arbitrairement avec [:len(ecg_emb)] (qui peut associer
    # un label à un patient différent de celui de l'embedding correspondant).
    df_test, _ = load_metadata_aligned(len(ecg_emb))

    if "lvef_lte_45_flag" not in df_test.columns:
        print("[SKIP] Clé 'lvef_lte_45_flag' non trouvée dans les métadonnées de Test.")
        return None

    y_all = df_test["lvef_lte_45_flag"].astype(int).values

    # -------------------------------------------------------------------------
    # 0. SPLIT STRICT DES DONNÉES (ÉVITE LE DATA LEAKAGE)
    # -------------------------------------------------------------------------
    indices = np.arange(len(y_all))
    idx_tr, idx_te = train_test_split(
        indices, test_size=0.3, random_state=42, stratify=y_all
    )
    ecg_tr, ecg_te = ecg_emb[idx_tr], ecg_emb[idx_te]
    graph_tr, graph_te = graph_emb[idx_tr], graph_emb[idx_te]
    tab_tr, tab_te = tab_scaled[idx_tr], tab_scaled[idx_te]
    y_tr, y_te = y_all[idx_tr], y_all[idx_te]

    # -------------------------------------------------------------------------
    # 1. PRÉDICTIONS DU MODÈLE FINAL (Phase 4 : ECG + Graph, 2 ENTRÉES SEULEMENT)
    #    -> tab_te n'est PAS passé, ce modèle n'a que 2 entrées.
    # -------------------------------------------------------------------------
    mlp_path = find_file(PATH_MLP, filename="best_mlp_phase4_ecg_graph_f2.keras")
    mlp_full = tf.keras.models.load_model(mlp_path, compile=False, safe_mode=False)
    raw_full = mlp_full.predict([ecg_te, graph_te], verbose=0).flatten()
    calibrator = pipeline_config.get("calibrator")
    calib_full = calibrator.predict(raw_full) if calibrator else raw_full

    # -------------------------------------------------------------------------
    # 2. ENTRAÎNEMENT DE LA CONFIG 1 : ECG SEUL (Entraîné sur TRAIN, Évalué sur TEST)
    # -------------------------------------------------------------------------
    model_ecg = build_baseline_mlp_ecg_only(ecg_tr.shape[1])
    model_ecg.fit(ecg_tr, y_tr, epochs=15, batch_size=32, verbose=0, validation_split=0.1)
    pred_ecg = model_ecg.predict(ecg_te, verbose=0).flatten()

    # -------------------------------------------------------------------------
    # 3. ENTRAÎNEMENT DE LA CONFIG 2 : ECG + TAB RÉINJECTÉ (Entraîné sur TRAIN, Évalué sur TEST)
    # -------------------------------------------------------------------------
    model_ecg_tab = build_baseline_mlp_ecg_tab(ecg_tr.shape[1], tab_tr.shape[1])
    model_ecg_tab.fit([ecg_tr, tab_tr], y_tr, epochs=15, batch_size=32, verbose=0, validation_split=0.1)
    pred_ecg_tab = model_ecg_tab.predict([ecg_te, tab_te], verbose=0).flatten()

    # -------------------------------------------------------------------------
    # DICTIONNAIRE DES RÉSULTATS D'ABLATION (noms corrigés, sans "Tab" trompeur en #3)
    # -------------------------------------------------------------------------
    models_dict = {
        "1. ECG (tab fusionné implicitement, Phase 2)": pred_ecg,
        "2. ECG + Tab réinjecté explicitement (baseline ad hoc)": pred_ecg_tab,
        "3. ECG + Graph (modèle final, Phase 4)": calib_full,
    }

    # -------------------------------------------------------------------------
    # CALCUL DES MÉTRIQUES ET COURBES SUR LE TEST SET UNIQUEMENT
    # -------------------------------------------------------------------------
    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    ablation_metrics = []
    for name, preds in models_dict.items():
        auc = roc_auc_score(y_te, preds)
        brier = brier_score_loss(y_te, preds)

        fpr, tpr, thresholds = roc_curve(y_te, preds)
        idx_recall_80 = np.argmin(np.abs(tpr - 0.80))
        spec_at_80_rec = 1.0 - fpr[idx_recall_80]

        ablation_metrics.append({
            "Configuration / Modèle": name,
            "AUC-ROC": round(auc, 4),
            "Brier Score": round(brier, 4),
            "Spécificité (à Rappel=80%)": f"{spec_at_80_rec*100:.1f}%",
        })

        axes[0].plot(fpr, tpr, label=f"{name} (AUC = {auc:.3f})", linewidth=2)
        prob_true, prob_pred = calibration_curve(y_te, preds, n_bins=10, strategy='uniform')
        axes[1].plot(prob_pred, prob_true, marker='o', label=f"{name} (Brier = {brier:.3f})", linewidth=2)

    axes[0].plot([0, 1], [0, 1], 'k--', alpha=0.6, label="Aléatoire (AUC = 0.500)")
    axes[0].set_xlabel("1 - Spécificité (Taux Faux Positifs)")
    axes[0].set_ylabel("Sensibilité / Rappel (Taux Vrais Positifs)")
    axes[0].set_title("Comparaison ROC-AUC : fusion implicite vs réinjection explicite vs graph")
    axes[0].legend(loc="lower right", fontsize=8)
    axes[0].grid(True, alpha=0.3)

    axes[1].plot([0, 1], [0, 1], 'k--', label="Calibration Parfaite")
    axes[1].set_xlabel("Probabilité Prédite Moyenne")
    axes[1].set_ylabel("Fraction de Positifs Réels")
    axes[1].set_title("Ablation Study : Gain en Calibration (Fiabilité)")
    axes[1].legend(loc="upper left", fontsize=8)
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    df_ablation = pd.DataFrame(ablation_metrics)
    print("\n" + "=" * 60)
    print(" TABLEAU COMPARATIF D'ABLATION (À PRÉSENTER AU JURY)")
    print("=" * 60)
    print(df_ablation.to_string(index=False))
    print("\nNote méthodologique : le tabulaire est fusionné une seule fois, en amont,")
    print("lors de la production de l'embedding ECG (Phase 2). Ni la Phase 3 (graph")
    print("enrichment) ni le modèle final (Phase 4) ne le réinjectent explicitement.")
    print("Seule la config 2 ci-dessus le réinjecte, à titre comparatif uniquement.")
    return df_ablation

# ============================== 14. EXÉCUTION PRINCIPALE =======================
def main():
    # --- Chargement du modèle et des données test ---
    atlas_h5_path = find_file(PATH_ATLAS_H5, filename="UKBRVLV.h5")
    pipeline_config = load_pipeline_config()

    # --- Étude d'ablation (apport du Graph Enrichment) ---
    run_ablation_study(pipeline_config)

    ecg_emb, graph_emb, tab_scaled = load_ecg_derived_data()
    raw_prob, lvef_prob = predict_lvef_probabilities(ecg_emb, graph_emb, tab_scaled, pipeline_config)

    # >>> FIX ALIGNEMENT METADATA / EMBEDDINGS <<<
    # On réconcilie df_test avec la longueur RÉELLE de ecg_emb (n_expected),
    # exactement comme le fait app.py via load_metadata_aligned(). Sans ce
    # fix, "index #8" ici (Kaggle) et "index #8" dans Streamlit peuvent
    # désigner deux patients physiquement différents si le CSV filtré
    # (most_recent_ecg==1) et le CSV complet ont des tailles différentes
    # pour le split test — ce qui produit des probabilités affichées
    # différentes pour le "même" index sans que le modèle soit en cause.
    df_test, id_col = load_metadata_aligned(len(ecg_emb))
    tab_raw = df_test[TAB_FEATURE_NAMES].values.astype(float)

    raw_waveforms_test = np.load(PATH_RAW_WAVEFORMS_TEST, mmap_mode="r")
    assert raw_waveforms_test.shape[0] == len(df_test), (
        f"Désalignement: raw_waveforms_test a {raw_waveforms_test.shape[0]} patients, "
        f"df_test en a {len(df_test)} — vérifier l'ordre du split test."
    )
    healthy_vols = compute_healthy_volumes(atlas_h5_path)

    # --- Patient aléatoire : vue 3D statique + jumeau 4D ---
    rand_idx = random.randint(0, len(lvef_prob) - 1)
    patient_id = df_test.iloc[rand_idx][id_col] if id_col in df_test.columns else f"Idx_{rand_idx}"
    print(f"🎲 Patient sélectionné aléatoirement : {patient_id} (Index {rand_idx})")

    res = visualize_patient(
        atlas_h5_path=atlas_h5_path, ecg_emb=ecg_emb, graph_emb=graph_emb, tab_scaled=tab_scaled,
        tab_raw=tab_raw, lvef_probabilities=lvef_prob, healthy_vols=healthy_vols,
        patient_idx=rand_idx, title_suffix=f" — Patient {patient_id}",
    )
    print(f"• Probabilité LVEF ≤ 45% : {res['lvef_probability']*100:.1f}%")
    print(f"• Durée QRS : {res['qrs_duration_ms']:.1f} ms | Fréquence Cardiaque : {res['heart_rate_bpm']:.1f} bpm")
    print(f"• Variation VG / VD vs Sain : {res['volume_pct_change']['LV']:+.1f}% / {res['volume_pct_change']['RV']:+.1f}%\n")

    plot_full_ecg_verification(raw_waveforms_test, df_test, id_col, tab_raw, rand_idx, fs=250)
    create_dynamic_digital_twin(
        atlas_h5_path=atlas_h5_path, patient_idx=rand_idx, patient_id_label=patient_id,
        lvef_probabilities=lvef_prob, tab_raw=tab_raw, raw_waveforms=raw_waveforms_test,
        n_frames=20, save_path=f"/kaggle/working/jumeau_4D_patient_{patient_id}.gif",
    )

    # --- 3 patients représentatifs (sain / IC modérée / IC élevée) ---
    patients_par_severite = select_patients_by_probability_thresholds(lvef_prob, PROBABILITY_THRESHOLDS)
    print("Patients sélectionnés par seuil de probabilité :")
    for label, info in patients_par_severite.items():
        pid = df_test.iloc[info["index"]][id_col] if id_col in df_test.columns else f"Idx_{info['index']}"
        low, high = info["seuils"]
        print(f"  • {SEVERITY_LABELS_FR[label]:<18} (seuils [{low:.2f}, {high:.2f}[) -> Patient {pid} "
              f"(index #{info['index']}, proba LVEF≤45% = {info['lvef_probability']*100:.1f}%)")

    shared_max_deform, max_deform_par_patient = compute_shared_max_deformation(
        atlas_h5_path, patients_par_severite, lvef_prob, tab_raw
    )
    print("\n--- ÉCHELLE DE COULEUR COMMUNE ---")
    for label, val in max_deform_par_patient.items():
        print(f"  • {SEVERITY_LABELS_FR[label]:<18} -> déformation max individuelle = {val:.1f} mm")
    print(f"  ==> max_deform PARTAGÉ appliqué aux 3 animations : {shared_max_deform:.1f} mm\n")

    animations_generees = {}
    for label, info in patients_par_severite.items():
        idx = info["index"]
        pid = df_test.iloc[idx][id_col] if id_col in df_test.columns else f"Idx_{idx}"
        label_complet = f"{pid} — {SEVERITY_LABELS_FR[label]}"
        save_path = f"/kaggle/working/jumeau_4D_{label}_{pid}.gif"

        print(f"\n--- Vérification ECG pour {SEVERITY_LABELS_FR[label]} ({pid}) ---")
        plot_full_ecg_verification(raw_waveforms_test, df_test, id_col, tab_raw, idx, fs=250,
                                    save_path=f"/kaggle/working/ecg_verif_{label}_{pid}.png")

        create_dynamic_digital_twin(
            atlas_h5_path=atlas_h5_path, patient_idx=idx, patient_id_label=label_complet,
            lvef_probabilities=lvef_prob, tab_raw=tab_raw, raw_waveforms=raw_waveforms_test,
            n_frames=20, save_path=save_path, show_deformation=True, blend_weight=0.6,
            max_deform=shared_max_deform, cmap_name="inferno",
            smoothing_iterations=5, smoothing_lambda=0.6,
        )
        animations_generees[label] = save_path

    print("\nAnimations 4D générées (échelle de couleur commune) :")
    for label, path in animations_generees.items():
        print(f"  • {SEVERITY_LABELS_FR[label]:<18} -> {path}")

    # --- Figure de comparaison finale (ED) ---
    plot_comparisonvf_3_patients(
        atlas_h5_path=atlas_h5_path, patients_par_severite=patients_par_severite,
        lvef_prob=lvef_prob, tab_raw=tab_raw, df_test=df_test, id_col=id_col,
        shared_max_deform=shared_max_deform, max_deform_par_patient=max_deform_par_patient,
        phase="ED", save_path="/kaggle/working/comparaison_3_vf_patients_ED.png",
    )


if __name__ == "__main__":
    main()
```

```python
%%writefile /kaggle/working/app.py
import os
import sys
import io
from typing import Dict, Optional, List

_APP_DIR = os.path.dirname(os.path.abspath(__file__))
if _APP_DIR not in sys.path:
    sys.path.insert(0, _APP_DIR)

import numpy as np
import pandas as pd
import joblib
import tensorflow as tf

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from matplotlib.colors import Normalize, to_rgb
from matplotlib.patches import Patch
from mpl_toolkits.mplot3d.art3d import Poly3DCollection, Line3DCollection

import streamlit as st
from scipy import sparse
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
from functools import lru_cache

try:
    from ukb import atlas as ukb_atlas
    from ukb import surface as ukb_surface
except ImportError:
    ukb_atlas = None
    ukb_surface = None

KAGGLE_INPUT = "/kaggle/input/datasets"

PATH_MLP         = os.path.join(KAGGLE_INPUT, "nadaelkamel3/phase4res-avecwl/best_mlp_phase4_ecg_graph_f2.keras")
PATH_PIPELINE    = os.path.join(KAGGLE_INPUT, "nadaelkamel3/phase4res-avecwl/echonext_pipeline_config.pkl")
PATH_ATLAS_H5    = os.path.join(KAGGLE_INPUT, "nadaelkamel3/ukbrvlv/UKBRVLV.h5")
PATH_METADATA    = os.path.join(KAGGLE_INPUT, "nadakamel3/echonext-metada-preprocessed/echonext_metadata_clean.csv")

PATH_EMB_TEST    = os.path.join(KAGGLE_INPUT, "nadakamel3/phase2-wl/embeddings_test.npy")
PATH_DGRAPH_TEST = os.path.join(KAGGLE_INPUT, "nadakamel3/embeddingsenrichis-w-tf1/d_graph_test.npy")
PATH_WAVE_TEST   = os.path.join(KAGGLE_INPUT, "nadakamel3/echonext-waveforms-preprocessed/EchoNext_test_cleaned.npy")

TAB_FEATURE_NAMES = ["sex", "ventricular_rate", "atrial_rate", "pr_interval", "qrs_duration", "qt_corrected", "age_at_ecg"]

TARGET_FS = 250

PHYSIO_BOUNDS = {0: {"min": -2.8, "max": +1.5}, 1: {"min": 0.0, "max": 2.5}}

DEFAULT_TAB_VALUES = {
    "sex": 1,
    "ventricular_rate": 70.0,
    "atrial_rate": 70.0,
    "pr_interval": 160.0,
    "qrs_duration": 90.0,
    "qt_corrected": 400.0,
    "age_at_ecg": 55.0,
}

MESH_STRUCTURES = {
    "EPI":  dict(color="lightgray",  alpha=0.15, label="Épicarde"),
    "LV":   dict(color="firebrick",  alpha=0.95, label="Ventricule gauche (VG)"),
    "RV":   dict(color="steelblue",  alpha=0.95, label="Ventricule droit (VD)"),
    "RVFW": dict(color="royalblue",  alpha=0.55, label="Paroi libre VD"),
    "MV":   dict(color="gold",       alpha=0.95, label="Valve mitrale"),
    "AV":   dict(color="darkorange", alpha=0.95, label="Valve aortique"),
    "TV":   dict(color="khaki",      alpha=0.95, label="Valve tricuspide"),
    "PV":   dict(color="orange",     alpha=0.95, label="Valve pulmonaire"),
}
_DEFORMATION_STRUCTURES = {"LV", "RV", "RVFW"}
_CHAMBER_NAMES = {"LV", "RV", "RVFW"}
_VALVE_NAMES   = {"MV", "AV", "TV", "PV"}


# ═══════════════════════════════════════════════════════════════════════════════
# CONSEILS CLINIQUES SIMPLIFIÉS — VERBES IMPÉRATIFS
# ═══════════════════════════════════════════════════════════════════════════════
CLINICAL_ADVICE = {
    "sain": {
        "emoji": "✅",
        "color": "#22c55e",
        "title": "Profil cardiaque dans les normes",
        "summary": "Aucun signe de dysfonction VG détecté. Le jumeau numérique ne montre pas d'écart significatif.",
        "prob_range": "< 15 %",
        "bilan": [
            "**Réaliser** un ECG de repos annuel si ≥ 50 ans ou antécédents familiaux.",
            "**Contrôler** la pression artérielle à chaque consultation (objectif < 130/80 mmHg).",
            "**Prescrire** un bilan lipidique annuel (LDL selon le risque SCORE2).",
            "**Dépister** la glycémie à jeun tous les 3 ans si ≥ 40 ans ou surpoids.",
        ],
        "analyses": [
            "**Vérifier** le bilan lipidique complet (LDL, HDL, TG, CT).",
            "**Dosage** de la créatinine + DFG si facteurs de risque rénaux.",
            "**Glycémie à jeun** ou HbA1c selon les facteurs de risque métaboliques.",
        ],
        "recommandations": [
            "**Prescrire** ≥ 150 min/semaine d'activité physique d'endurance modérée.",
            "**Conseiller** un régime méditerranéen, sel < 5 g/jour, arrêt tabagique.",
            "**Limiter** la consommation d'alcool (≤ 2 verres/jour pour l'homme, ≤ 1 pour la femme).",
            "**Maintenir** un IMC entre 18.5 et 25 kg/m².",
        ],
        "surveillance": [
            "**Surveiller** l'apparition de dyspnée d'effort, d'œdèmes ou de palpitations.",
            "**Peser** le patient quotidiennement si antécédent familial d'IC.",
            "**Consulter** rapidement en cas de gain de poids > 2 kg en 3 jours.",
        ],
        "alert": None,
        "referral": "Aucune orientation spécialisée requise. Suivi par le médecin traitant.",
    },
    "moderee": {
        "emoji": "⚠️",
        "color": "#f59e0b",
        "title": "Suspicion d'insuffisance cardiaque modérée",
        "summary": "Probabilité intermédiaire de dysfonction VG (LVEF ≤ 45 %). Écarts modérés sur le jumeau numérique 3D.",
        "prob_range": "15 – 50 %",
        "bilan": [
            "**Prescrire** une échocardiographie Doppler de repos en priorité (délai < 4 semaines).",
            "**Réaliser** un ECG de 24–72 h (Holter) en cas de palpitations ou dyspnée paroxystique.",
            "**Demander** une radiographie thoracique (recherche de redistribution vasculaire ou épanchement pleural).",
            "**Évaluer** la classe NYHA et le score de dyspnée (mMRC).",
        ],
        "analyses": [
            "**Dosage urgent** du NT-proBNP (ou BNP) — valeur > 400 pg/mL = IC décompensée probable.",
            "**Prescrire** NFS, créatinine + DFG, ionogramme (recherche d'hyponatrémie).",
            "**Ajouter** ASAT/ALAT, TSH, glycémie à jeun, HbA1c.",
            "**Rechercher** une anémie (NFS) et évaluer la ferritinémie si fatigue marquée.",
        ],
        "recommandations": [
            "**Optimiser** les doses d'IEC / ARA2 / ARNI si déjà traité (titération progressive).",
            "**Vérifier** l'observance du bêtabloquant cardiosélectif et du MRA.",
            "**Prescrire** un diurétique (furosémide ou HCTZ) si œdèmes ou crépitants bilatéraux.",
            "**Envisager** d'emblée un SGLT2i (dapagliflozine ou empagliflozine) — bénéfice prouvé même sans diabète.",
            "**Discuter** l'anticoagulation si FA détectée (score CHA₂DS₂-VASc ≥ 2).",
        ],
        "surveillance": [
            "**Programmer** une consultation cardiologie sous 2–4 semaines avec les résultats.",
            "**Surveiller** le poids quotidien + restriction hydrosodée (sel < 3 g/j, liquides < 1.5–2 L/j).",
            "**Contrôler** la kaliémie et la fonction rénale à J3, J7 puis mensuellement sous IEC/ARA2 + MRA.",
            "**Réévaluer** la clinique sous 4–6 semaines après chaque ajustement thérapeutique.",
        ],
        "alert": "30–40 % des patients à risque intermédiaire évoluent vers une IC déclarée sans prise en charge précoce.",
        "referral": "**Orienter** vers cardiologie ou consultation d'insuffisance cardiaque sous 2–4 semaines.",
    },
    "elevee": {
        "emoji": "🚨",
        "color": "#ef4444",
        "title": "Risque élevé de dysfonction VG sévère",
        "summary": "Fortes probabilités de LVEF ≤ 45 %. Déformations majeures du jumeau numérique 3D (ED/ES). QRS probablement ≥ 120 ms.",
        "prob_range": "> 50 %",
        "bilan": [
            "**Réaliser** une échocardiographie Doppler en urgence (délai < 48–72 h).",
            "**Prescrire** un ECG 12 dérivations complet (recherche de BBG, BBD, signes d'ischémie).",
            "**Demander** une radiographie thoracique d'urgence (index cardiaque, congestion, épanchement).",
            "**Programmer** une IRM cardiaque si étiologie indéterminée après écho (fibrose, myocardite, CM).",
            "**Discuter** une coronarographie si suspicion d'étiologie ischémique.",
        ],
        "analyses": [
            "**Dosage immédiat** du NT-proBNP (BNP > 2000 pg/mL = hospitalisation recommandée).",
            "**Prescrire** NFS, créatinine + DFG CKD-EPI, ionogramme complet (Na, K, Cl, bicarbonates).",
            "**Ajouter** TSH, ferritinémie + transferrine, ASAT/ALAT/bilirubine, glycémie.",
            "**Réaliser** une gazométrie artérielle si dyspnée au repos ou SpO₂ < 94 %.",
            "**Rechercher** une acidose lactique si état de choc ou hypoperfusion périphérique.",
        ],
        "recommandations": [
            "**Initier** la quadrithérapie de guideline : ARNI (ou IEC/ARA2) + bêtabloquant + MRA + SGLT2i.",
            "**Administrer** du furosémide IV si congestion aiguë (crépitants, œdèmes, BNP élevé).",
            "**Anticoaguler** systématiquement si FA ou thrombus intracavitaire à l'écho (AOD de préférence).",
            "**Évaluer** l'indication de CRT-D si QRS ≥ 150 ms + BBG + FEVG ≤ 35 %.",
            "**Discuter** un DAI en prévention primaire après 3 mois d'optimisation médicale.",
            "**Envisager** la transplantation cardiaque ou un LVAD en cas de choc cardiogénique réfractaire.",
        ],
        "surveillance": [
            "**Hospitaliser** si NYHA III–IV, congestion aiguë, œdèmes généralisés ou BNP > 2000 pg/mL.",
            "**Surveiller** en continu : PA, diurèse, SatO₂, ionogramme (risque d'hyperkaliémie sous MRA).",
            "**Programmer** une consultation cardiologie sous 48 h si pas d'hospitalisation.",
            "**Réévaluer** à J7, J30, M3 : clinique, biologie (créatinine, kaliémie), écho de contrôle à M3–M6.",
            "**Discuter** en RCP / MDT si étiologie rare, cas complexe ou candidature à greffe.",
        ],
        "alert": "Profil à haut risque d'événements cardiovasculaires majeurs : hospitalisation, mort subite, IC réfractaire.",
        "referral": "**Orienter d'urgence** vers cardiologie sous 48 h. Si dyspnée au repos ou congestion aiguë → urgences immédiates.",
    },
}


def find_file(expected_path: str, search_root: str = "/kaggle/input", filename: Optional[str] = None) -> str:
    if os.path.exists(expected_path):
        return expected_path
    target = filename or os.path.basename(expected_path)
    for root, _, files in os.walk(search_root):
        for f in files:
            if f.lower() == target.lower():
                return os.path.join(root, f)
    raise FileNotFoundError(f"{target} introuvable sous {search_root}")


@st.cache_resource
def load_phase4_mlp():
    p = find_file(PATH_MLP, filename="best_mlp_phase4_ecg_graph_f2.keras")
    return tf.keras.models.load_model(p, compile=False, safe_mode=False)

@st.cache_resource
def load_pipeline_config() -> dict:
    try:
        p = find_file(PATH_PIPELINE, filename="echonext_pipeline_config.pkl")
        return joblib.load(p)
    except Exception:
        return {"calibrator": None, "threshold_recall": 0.5}

@st.cache_data
def load_metadata_aligned(n_expected: int):
    path = find_file(PATH_METADATA, filename="echonext_metadata_clean.csv")
    df = pd.read_csv(path)

    id_col = next((c for c in ["patient_key", "eid", "patient_id", "subject_id", "id"] if c in df.columns), None)
    if id_col is None:
        df["_temp_id"] = df.index.astype(str)
        id_col = "_temp_id"

    split_col = next((c for c in ("split", "dataset_split", "set", "fold") if c in df.columns), None)
    if split_col is None:
        raise ValueError("Colonne de split introuvable dans les métadonnées.")

    if "most_recent_ecg" in df.columns:
        df_filtre = df[df["most_recent_ecg"] == 1]
        test_filtre = df_filtre[df_filtre[split_col].astype(str).str.lower() == "test"].reset_index(drop=True)
        test_complet = df[df[split_col].astype(str).str.lower() == "test"].reset_index(drop=True)

        if len(test_filtre) == n_expected:
            return test_filtre, id_col
        elif len(test_complet) == n_expected:
            return test_complet, id_col
        else:
            raise ValueError(
                f"[DÉSALIGNEMENT METADATA / EMBEDDINGS] "
                f"embeddings_test.npy = {n_expected} lignes | "
                f"CSV filtré (most_recent_ecg==1, split=test) = {len(test_filtre)} lignes | "
                f"CSV complet (split=test) = {len(test_complet)} lignes. "
                f"Impossible de garantir la correspondance patient <-> embedding — "
                f"vérifiez la source des .npy ou du CSV avant de continuer."
            )
    else:
        df_test = df[df[split_col].astype(str).str.lower() == "test"].reset_index(drop=True)
        if len(df_test) != n_expected:
            raise ValueError(
                f"[DÉSALIGNEMENT METADATA / EMBEDDINGS] "
                f"{len(df_test)} lignes de métadonnées vs {n_expected} lignes d'embeddings."
            )
        return df_test, id_col

@st.cache_data
def load_test_arrays():
    ecg_emb = np.load(find_file(PATH_EMB_TEST, filename="embeddings_test.npy"))
    graph_emb = np.load(find_file(PATH_DGRAPH_TEST, filename="d_graph_test.npy"))
    return ecg_emb, graph_emb


def phase4_predict(ecg_emb: np.ndarray, graph_emb: np.ndarray, cfg: dict) -> float:
    mlp = load_phase4_mlp()
    raw = mlp.predict([
        np.expand_dims(ecg_emb, 0),
        np.expand_dims(graph_emb, 0),
    ], verbose=0).flatten()[0]
    cal = cfg.get("calibrator")
    if cal is not None:
        proba = float(cal.predict(np.array([raw]))[0])
    else:
        proba = float(raw)
    return float(np.clip(proba, 0.001, 0.999))


# ═══════════════════════════════════════════════════════════════════════════════
# JUMEAU NUMÉRIQUE 3D BIDIRECTIONNEL (ED ↔ ES)
# ═══════════════════════════════════════════════════════════════════════════════
def mode_weights_to_score(mode_weights: dict, n_modes: Optional[int] = None) -> np.ndarray:
    if n_modes is None:
        n_modes = max(mode_weights.keys(), default=-1) + 1
    score = np.zeros(max(n_modes, 1))
    for idx, w in mode_weights.items():
        score[idx] = w
    return score

def get_structure_mesh(name: str, points: np.ndarray):
    if name == "EPI":
        return ukb_surface.get_epi_mesh(points=points)
    if name in _CHAMBER_NAMES:
        return ukb_surface.get_chamber_mesh(surface_name=name, points=points)
    if name in _VALVE_NAMES:
        return ukb_surface.get_valve_mesh(surface_name=name, points=points)
    raise ValueError(f"Structure inconnue : {name}")

@lru_cache(maxsize=128)
def _build_biv_surfaces_cached(atlas_h5_path: str, weights_key: tuple) -> Dict[str, Dict[str, object]]:
    mode_weights = dict(weights_key)
    score = mode_weights_to_score(mode_weights)
    points = ukb_atlas.generate_points(filename=atlas_h5_path, score=score)
    return {
        case_name: {name: get_structure_mesh(name, pts) for name in MESH_STRUCTURES}
        for case_name, pts in [("ED", points.ED), ("ES", points.ES)]
    }

def build_biv_surfaces(atlas_h5_path: str, mode_weights: dict) -> Dict[str, Dict[str, object]]:
    weights_key = tuple(sorted(mode_weights.items()))
    return _build_biv_surfaces_cached(atlas_h5_path, weights_key)

def compute_mode_weights(lvef_probability: float, qrs_duration_ms: float,
                          qrs_normal_upper: float = 110.0, qrs_pathological: float = 120.0,
                          max_std: float = 3.0) -> dict:
    w0_raw = -np.clip(lvef_probability * max_std, 0.0, max_std)
    w0 = np.clip(w0_raw, PHYSIO_BOUNDS[0]["min"], PHYSIO_BOUNDS[0]["max"])

    if qrs_duration_ms >= qrs_pathological:
        qrs_excess = qrs_duration_ms - qrs_pathological
        w1_raw = np.clip(qrs_excess / 20.0, 0.0, max_std)
    elif qrs_duration_ms >= qrs_normal_upper:
        w1_raw = 0.3 * (qrs_duration_ms - qrs_normal_upper) / (qrs_pathological - qrs_normal_upper)
    else:
        w1_raw = 0.0
    w1 = np.clip(w1_raw, PHYSIO_BOUNDS[1]["min"], PHYSIO_BOUNDS[1]["max"])
    return {0: w0, 1: w1}

def compute_vertex_deformation(mesh_patient, mesh_healthy) -> np.ndarray:
    return np.linalg.norm(mesh_patient.points - mesh_healthy.points, axis=1)

def build_laplacian_smoother(faces: np.ndarray, n_vertices: int):
    a, b, c = faces[:, 0], faces[:, 1], faces[:, 2]
    rows = np.concatenate([a, b, c, b, c, a])
    cols = np.concatenate([b, c, a, a, b, c])
    data = np.ones(len(rows))
    A = sparse.coo_matrix((data, (rows, cols)), shape=(n_vertices, n_vertices)).tocsr()
    deg = np.asarray(A.sum(axis=1)).flatten()
    deg[deg == 0] = 1.0
    return sparse.diags(1.0 / deg) @ A

def smooth_scalar_field(W, values: np.ndarray, n_iter: int = 5, lam: float = 0.6) -> np.ndarray:
    smoothed = values.astype(float).copy()
    for _ in range(n_iter):
        smoothed = (1.0 - lam) * smoothed + lam * (W @ smoothed)
    return smoothed

_LAPLACIAN_CACHE: Dict[tuple, sparse.spmatrix] = {}

def get_cached_laplacian(name: str, phase: str, mesh):
    key = (name, phase)
    if key not in _LAPLACIAN_CACHE:
        faces = mesh.cells[0].data
        _LAPLACIAN_CACHE[key] = build_laplacian_smoother(faces, len(mesh.points))
    return _LAPLACIAN_CACHE[key]

def shade_triangles(tris: np.ndarray, base_color: str, alpha: float,
                     light_direction=(-1.0, -1.0, 1.2), ambient: float = 0.35, diffuse: float = 0.65) -> np.ndarray:
    base = np.array(to_rgb(base_color))
    v0, v1, v2 = tris[:, 0], tris[:, 1], tris[:, 2]
    normals = np.cross(v1 - v0, v2 - v0)
    norm_len = np.linalg.norm(normals, axis=1, keepdims=True)
    norm_len[norm_len == 0] = 1.0
    normals /= norm_len
    light = np.asarray(light_direction, dtype=float)
    light /= np.linalg.norm(light)
    intensity = np.clip(np.abs(normals @ light), 0, 1)
    shade = ambient + diffuse * intensity
    rgb = np.clip(base[None, :] * shade[:, None], 0, 1)
    return np.concatenate([rgb, np.full((rgb.shape[0], 1), alpha)], axis=1)

def _mesh_triangles(mesh):
    try:
        faces = np.asarray(mesh.cells[0].data)[:, :3].astype(int)
    except Exception:
        return None
    return np.asarray(mesh.points)[faces], faces

def _mesh_wireframe_segments(mesh):
    result = _mesh_triangles(mesh)
    if result is None:
        return None
    points, faces = np.asarray(mesh.points), result[1]
    edges = np.concatenate([faces[:, [0, 1]], faces[:, [1, 2]], faces[:, [2, 0]]], axis=0)
    edges = np.unique(np.sort(edges, axis=1), axis=0)
    return points[edges]

def render_biv_mesh(ax, surfaces_patient: Dict[str, object],
                     deformation_by_structure: Optional[Dict[str, np.ndarray]] = None,
                     blend_weight: float = 0.6, max_deform: float = 8.0, cmap_name: str = "inferno",
                     surfaces_reference: Optional[Dict[str, object]] = None,
                     reference_structures=None,
                     reference_color: str = "white", reference_linewidth: float = 0.7,
                     reference_alpha: float = 0.45, show_reference_label: bool = True,
                     title: str = "", title_color: str = "white", view_bounds: Optional[tuple] = None,
                     render_structures: Optional[set] = None):
    ax.clear()
    norm = Normalize(vmin=0.0, vmax=max_deform)
    cmap = matplotlib.colormaps[cmap_name]
    all_points = []

    for name, style in MESH_STRUCTURES.items():
        if render_structures is not None and name not in render_structures:
            continue
        mesh = surfaces_patient.get(name)
        if mesh is None:
            continue
        tris_faces = _mesh_triangles(mesh)
        if tris_faces is None:
            continue
        tris, faces = tris_faces
        rgba = shade_triangles(tris, style["color"], style["alpha"])

        has_deform = (deformation_by_structure is not None and name in _DEFORMATION_STRUCTURES and name in deformation_by_structure)
        if has_deform:
            tri_deform = deformation_by_structure[name][faces].mean(axis=1)
            overlay = cmap(norm(tri_deform))
            overlay[:, 3] = rgba[:, 3]
            rgba = (1.0 - blend_weight) * rgba + blend_weight * overlay

        ax.add_collection3d(Poly3DCollection(tris, facecolor=rgba, edgecolor="none"))
        all_points.append(mesh.points)

    if surfaces_reference is not None:
        structures = list(reference_structures) if reference_structures else list(surfaces_reference)
        segments = []
        for name in structures:
            mesh = surfaces_reference.get(name)
            if mesh is None:
                continue
            seg = _mesh_wireframe_segments(mesh)
            if seg is not None:
                segments.append(seg)
        if segments:
            wire = Line3DCollection(
                np.concatenate(segments, axis=0),
                colors=reference_color, linewidths=reference_linewidth,
                alpha=reference_alpha, linestyles="solid",
            )
            ax.add_collection3d(wire)
            if show_reference_label:
                ax.text2D(0.03, 0.04, "Contour blanc = référence saine",
                          transform=ax.transAxes, color="white", fontsize=8, alpha=0.75)

    if view_bounds is not None:
        (xmin, xmax), (ymin, ymax), (zmin, zmax) = view_bounds
    elif all_points:
        pts = np.concatenate(all_points, axis=0)
        mins, maxs = pts.min(axis=0), pts.max(axis=0)
        center = (mins + maxs) / 2.0
        half = (maxs - mins).max() / 2.0 * 1.12
        (xmin, xmax), (ymin, ymax), (zmin, zmax) = [(c - half, c + half) for c in center]
    else:
        (xmin, xmax), (ymin, ymax), (zmin, zmax) = (-60, 60), (-60, 60), (-60, 60)

    ax.set_xlim(xmin, xmax); ax.set_ylim(ymin, ymax); ax.set_zlim(zmin, zmax)
    ax.set_box_aspect([1, 1, 1])
    ax.set_title(title, fontsize=11, fontweight="bold", pad=5, color=title_color)
    ax.set_axis_off()

def compute_shared_view_bounds(surface_sets, padding_ratio: float = 0.12):
    all_pts = [mesh.points for surfaces in surface_sets for mesh in surfaces.values()]
    all_pts = np.concatenate(all_pts, axis=0)
    mins, maxs = all_pts.min(axis=0), all_pts.max(axis=0)
    center = (mins + maxs) / 2.0
    half_range = (maxs - mins).max() / 2.0 * (1.0 + padding_ratio)
    return tuple((c - half_range, c + half_range) for c in center)

def generate_digital_twin_figure(atlas_h5_path: str, prob: float, qrs_ms: float,
                                  max_deform: float = 8.0, cmap_name: str = "inferno",
                                  show_both_phases: bool = True):
    surfaces_healthy = build_biv_surfaces(atlas_h5_path, {})
    weights = compute_mode_weights(prob, qrs_ms)
    surfaces_patient = build_biv_surfaces(atlas_h5_path, weights)

    deform_by_structure = {"ED": {}, "ES": {}}
    for phase in ("ED", "ES"):
        for name in _DEFORMATION_STRUCTURES:
            mesh_pat = surfaces_patient[phase][name]
            mesh_hea = surfaces_healthy[phase][name]
            raw_def = compute_vertex_deformation(mesh_pat, mesh_hea)
            W = get_cached_laplacian(name, phase, mesh_pat)
            deform_by_structure[phase][name] = smooth_scalar_field(W, raw_def, n_iter=5, lam=0.6)

    view_bounds = compute_shared_view_bounds([
        surfaces_healthy["ED"], surfaces_patient["ED"],
        surfaces_healthy["ES"], surfaces_patient["ES"]
    ])

    if show_both_phases:
        fig = plt.figure(figsize=(20, 9), facecolor="#0d1117")

        # Grille personnalisée : 2 panneaux 3D + marge à droite pour la colorbar
        ax_ed = fig.add_axes([0.02, 0.08, 0.40, 0.82], projection="3d", facecolor="#0d1117")
        ax_es = fig.add_axes([0.44, 0.08, 0.40, 0.82], projection="3d", facecolor="#0d1117")

        render_biv_mesh(
            ax_ed, surfaces_patient["ED"],
            deformation_by_structure=deform_by_structure["ED"],
            blend_weight=0.6, max_deform=max_deform, cmap_name=cmap_name,
            surfaces_reference=surfaces_healthy["ED"],
            reference_structures=_DEFORMATION_STRUCTURES,
            reference_color="white", reference_linewidth=0.75, reference_alpha=0.50,
            title="Diastole (ED) — Remplissage", title_color="white",
            view_bounds=view_bounds,
        )
        render_biv_mesh(
            ax_es, surfaces_patient["ES"],
            deformation_by_structure=deform_by_structure["ES"],
            blend_weight=0.6, max_deform=max_deform, cmap_name=cmap_name,
            surfaces_reference=surfaces_healthy["ES"],
            reference_structures=_DEFORMATION_STRUCTURES,
            reference_color="white", reference_linewidth=0.75, reference_alpha=0.50,
            title="Systole (ES) — Éjection", title_color="white",
            view_bounds=view_bounds,
        )

        # Colorbar sur un axe dédié à droite — pas de chevauchement
        cax = fig.add_axes([0.87, 0.25, 0.015, 0.50])
        sm = cm.ScalarMappable(norm=Normalize(vmin=0.0, vmax=max_deform), cmap=cmap_name)
        sm.set_array([])
        cbar = fig.colorbar(sm, cax=cax)
        cbar.set_label("Écart vs réf. saine (mm)", color="white", fontsize=10, labelpad=12)
        cbar.ax.yaxis.set_tick_params(color="white", labelsize=9)
        plt.setp(plt.getp(cbar.ax.axes, "yticklabels"), color="white")

        # Légende en bas
        handles = [Patch(facecolor=s["color"], alpha=s["alpha"], label=s["label"]) for s in MESH_STRUCTURES.values()]
        fig.legend(handles=handles, loc="lower center", ncol=4, frameon=False, fontsize=9,
                   labelcolor="white", bbox_to_anchor=(0.45, 0.01))
    else:
        fig = plt.figure(figsize=(10, 10), facecolor="#0d1117")
        ax = fig.add_subplot(111, projection="3d", facecolor="#0d1117")
        render_biv_mesh(
            ax, surfaces_patient["ED"],
            deformation_by_structure=deform_by_structure["ED"],
            blend_weight=0.6, max_deform=max_deform, cmap_name=cmap_name,
            surfaces_reference=surfaces_healthy["ED"],
            reference_structures=_DEFORMATION_STRUCTURES,
            reference_color="white", reference_linewidth=0.75, reference_alpha=0.50,
            title="Jumeau Numérique 3D — Phase Diastole (ED)", title_color="white",
            view_bounds=view_bounds,
        )
        handles = [Patch(facecolor=s["color"], alpha=s["alpha"], label=s["label"]) for s in MESH_STRUCTURES.values()]
        ax.legend(handles=handles, loc="upper left", ncol=2, frameon=False, fontsize=8, labelcolor="white")

        sm = cm.ScalarMappable(norm=Normalize(vmin=0.0, vmax=max_deform), cmap=cmap_name)
        sm.set_array([])
        cbar = fig.colorbar(sm, ax=ax, shrink=0.45, aspect=18, pad=0.02)
        cbar.set_label("Écart vs référence saine (mm)", color="white", fontsize=9)
        cbar.ax.yaxis.set_tick_params(color="white")
        plt.setp(plt.getp(cbar.ax.axes, "yticklabels"), color="white")

    buf = io.BytesIO()
    fig.savefig(buf, format="png", dpi=150, bbox_inches="tight", facecolor="#0d1117")
    plt.close(fig)
    buf.seek(0)
    return buf


# ═══════════════════════════════════════════════════════════════════════════════
# INTERFACE STREAMLIT — CARDIOREPLICA SIMPLIFIÉ
# ═══════════════════════════════════════════════════════════════════════════════
def render_advice_panel(advice: dict, prob: float):
    """Affiche les recommandations cliniques simplifiées avec verbes impératifs."""
    color = advice["color"]
    emoji = advice["emoji"]

    # ── En-tête compact ──
    st.markdown(
        f"""
        <div style="padding:14px; border-radius:10px; background:{color}15; border-left:4px solid {color};">
            <div style="font-size:0.8em; font-weight:700; color:{color}; letter-spacing:0.5px;">
                PROFIL {advice['prob_range']}
            </div>
            <div style="font-size:1.1em; font-weight:700; color:#e6edf3; margin-top:4px;">
                {emoji} {advice['title']}
            </div>
            <p style="color:#9ca3af; margin:6px 0 10px 0; font-size:0.9em; line-height:1.4;">{advice['summary']}</p>
            <div style="display:flex; align-items:center; gap:8px;">
                <span style="color:#9ca3af; font-size:0.85em;">LVEF ≤ 45% :</span>
                <span style="font-weight:700; font-size:1.1em; color:{color};">{prob*100:.1f}%</span>
                <div style="flex:1; height:8px; background:#21262d; border-radius:4px; overflow:hidden; margin-left:6px;">
                    <div style="width:{min(prob*100,100)}%; height:100%; background:{color}; border-radius:4px;"></div>
                </div>
            </div>
        </div>
        """,
        unsafe_allow_html=True
    )

    # ── 4 onglets simplifiés ──
    tab_bilan, tab_analyses, tab_reco, tab_surv = st.tabs([
        "🔬 Bilan",
        "🧪 Analyses",
        "💊 Recommandations",
        "📡 Surveillance",
    ])

    with tab_bilan:
        st.markdown("**Examens et évaluations cliniques à réaliser :**")
        for item in advice["bilan"]:
            st.markdown(f"• {item}")

    with tab_analyses:
        st.markdown("**Biologie et examens complémentaires à prescrire :**")
        for item in advice["analyses"]:
            st.markdown(f"• {item}")

    with tab_reco:
        st.markdown("**Conduite à tenir et thérapeutique :**")
        for item in advice["recommandations"]:
            st.markdown(f"• {item}")

    with tab_surv:
        st.markdown("**Suivi et signes d'alerte :**")
        for item in advice["surveillance"]:
            st.markdown(f"• {item}")

    if advice["alert"]:
        st.warning(f"⚡ {advice['alert']}")

    st.info(f"🩺 **Orientation :** {advice['referral']}")


def main():
    st.set_page_config(page_title="CardioReplica AI", layout="wide", page_icon="🫀")

    st.markdown("""
    <style>
    .stTabs [data-baseweb="tab-list"] { gap: 6px; }
    .stTabs [data-baseweb="tab"] {
        background: #161b22;
        border-radius: 6px 6px 0 0;
        border: 1px solid #30363d;
        color: #9ca3af;
        padding: 8px 16px;
        font-weight: 600;
        font-size: 0.9em;
    }
    .stTabs [aria-selected="true"] {
        background: #21262d !important;
        color: #e6edf3 !important;
        border-bottom: 2px solid #ff3366 !important;
    }
    div[data-testid="stMetricValue"] { font-size: 1.5em; font-weight: 700; }
    </style>
    """, unsafe_allow_html=True)

    with st.sidebar:
        st.markdown("<h1 style='color:#ff3366;'>🫀 CardioReplica AI</h1>", unsafe_allow_html=True)
        st.markdown("**Jumeau numérique cardiaque 3D bidirectionnel** — Dépistage non-invasif de la dysfonction VG.")
        st.divider()
        st.markdown("""
        **Pipeline :**
        1. 📊 Embeddings ECG (Phase 2)
        2. 🕸️ Graph Enrichment (Phase 3)
        3. 🎯 Prédiction LVEF ≤ 45% (Phase 4)
        4. 🫀 Jumeau 3D ED/ES
        5. 💡 Aide à la décision clinique
        """)
        st.divider()
        st.caption("Mode démo sur patients réels du dataset test.")

    st.markdown("<h2 style='text-align:center; color:#e6edf3;'>🫀 CardioReplica — Dépistage & Aide à la Décision</h2>", unsafe_allow_html=True)
    st.markdown("<p style='text-align:center; color:#9ca3af; font-size:0.9em;'>ECG → IA → Jumeau 3D bidirectionnel → Conseils cliniques actionnables</p>", unsafe_allow_html=True)
    st.divider()

    ecg_emb, graph_emb = load_test_arrays()
    try:
        df_test, id_col = load_metadata_aligned(len(ecg_emb))
    except ValueError as e:
        st.error(f"❌ {e}")
        st.stop()

    st.info(f"Dataset test chargé : **{len(df_test)}** patients | Embeddings prêts.")

    idx = st.number_input("Index patient (0-based)", 0, len(df_test) - 1, 0, step=1)
    pid = df_test.iloc[idx][id_col] if id_col in df_test.columns else f"Idx_{idx}"

    if st.button("🔬 Analyser ce patient", use_container_width=True):
        with st.spinner(f"Analyse du patient {pid}..."):
            cfg = load_pipeline_config()
            ecg_e = ecg_emb[idx]
            graph_e = graph_emb[idx]

            prob = phase4_predict(ecg_e, graph_e, cfg)

            qrs_idx = TAB_FEATURE_NAMES.index("qrs_duration")
            qrs_raw = df_test.iloc[idx][TAB_FEATURE_NAMES[qrs_idx]] if TAB_FEATURE_NAMES[qrs_idx] in df_test.columns else np.nan
            if not np.isfinite(float(qrs_raw)):
                qrs_raw = DEFAULT_TAB_VALUES["qrs_duration"]
            qrs_ms = float(qrs_raw)

            if prob < 0.15:
                risk_key = "sain"
                risk_label = "Sain"
                color = "#22c55e"
                emoji = "✅"
            elif prob < 0.50:
                risk_key = "moderee"
                risk_label = "IC modérée"
                color = "#f59e0b"
                emoji = "⚠️"
            else:
                risk_key = "elevee"
                risk_label = "IC élevée"
                color = "#ef4444"
                emoji = "🚨"

            advice = CLINICAL_ADVICE[risk_key]

            col1, col2, col3 = st.columns(3)
            col1.metric("LVEF ≤ 45%", f"{prob*100:.1f}%")
            col2.metric("Risque", risk_label)
            col3.metric("QRS", f"{qrs_ms:.0f} ms")

            st.divider()

            st.subheader("🫀 Jumeau Numérique 3D Bidirectionnel")
            atlas_path = find_file(PATH_ATLAS_H5, filename="UKBRVLV.h5")
            twin_buf = generate_digital_twin_figure(atlas_path, prob, qrs_ms, max_deform=8.0, show_both_phases=True)
            st.image(twin_buf, caption="ED = Diastole (remplissage) | ES = Systole (éjection) | Contour blanc = référence saine", use_container_width=True)

            st.divider()

            st.subheader("📋 Recommandations Cliniques")
            render_advice_panel(advice, prob)

            st.divider()

            st.subheader("📡 ECG — Dérivation II")
            try:
                raw_wf = np.load(find_file(PATH_WAVE_TEST, filename="EchoNext_test_cleaned.npy"), mmap_mode="r")
                fig_ecg, ax_ecg = plt.subplots(figsize=(12, 3.2), facecolor="#0d1117")
                ax_ecg.set_facecolor("#0d1117")
                t = np.arange(raw_wf.shape[1]) / 250.0
                ax_ecg.plot(t, raw_wf[idx, :, 1], color="#00ffcc", lw=0.8)
                ax_ecg.set_title(f"Dérivation II — Patient {pid}", color="white", fontsize=11)
                ax_ecg.set_xlabel("Temps (s)", color="#9ca3af", fontsize=9)
                ax_ecg.set_ylabel("Amplitude", color="#9ca3af", fontsize=9)
                ax_ecg.tick_params(colors="white", labelsize=8)
                ax_ecg.grid(True, alpha=0.12, color="white")
                for spine in ax_ecg.spines.values():
                    spine.set_color("#30363d")
                buf_ecg = io.BytesIO()
                fig_ecg.savefig(buf_ecg, format="png", dpi=120, facecolor="#0d1117", bbox_inches="tight")
                plt.close(fig_ecg)
                buf_ecg.seek(0)
                st.image(buf_ecg, caption="Signal ECG (Dérivation II) — Amplitudes potentiellement normalisées (Z-score)", use_container_width=True)
            except Exception as e:
                st.warning(f"Signal ECG indisponible : {e}")


if __name__ == "__main__":
    main()
```
